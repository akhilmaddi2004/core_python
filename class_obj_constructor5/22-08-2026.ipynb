## Class, Object & Constructor

```python
# Class:
# Class is a blueprint of object
# we can create claa by using class-keyword
# A class is like an object constructor, or a "blueprint" for creating objects.

# syntax:
class class_name:
    pass # code of class block

# we can create custom functions, codes using class
```

```python
# object creation is the heart of class execution
# Object creation is equal to function calling

class A:
    pass
obj1 = A()
print(type(obj1))
```

```python
obj1 = A()
obj2 = A()
print(id(obj1))
print(id(obj2))
# Both object's memory locations are different because we are calling A() (not assigning A(class) to a variable/object) so new object will be created in new memory location

# m1 is a function f1,f2 are variables to used in function referencing
# In functions: f1 = m1, f2 = m1 are has same memory location because we are assigning m1 to f1 and f2
```

```python
# Object creation:
# __init__ and __new__ these both are helpful in object creation
# init is a constructor and also a method

# - __new__ method is called first in class
# - new memory space/location is created and allocated returns to __init__
# - __init__ loads/processes properties(instance)

# we have 2 types of variable in class
# 1. class variables/property:
# - These variables are class depended or related to class variables
# - the variables which are belong to class and not depend on object
class A:
    a = 20 # class property/variable
obj = A()
print(obj.a)
print(A.a)

# 2. Instance variables
# To use current object we use self.
# init's first parameter is object address(self)
class phone:
    software = "Android" # class property
    count = 0
    def __init__(self,brand,price):
        self.brand = brand
        self.price = price
        phone.count += 1
phone1 = phone("Samsung",156000)
phone2 = phone("IQOO",56000)
print(phone1.software)
print(phone1.brand)
print(phone1.price)
print()
print(phone2.software)
print(phone2.brand)
print(phone2.price)
print(phone.count)
```
