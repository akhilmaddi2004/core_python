## Modification of class variables


```python
class A:
    x = 20
    def __init__(self,y):
        self.y = y
# obj = A()
# TypeError: A.__init__() missing 1 required positional argument: 'y'

# class variables are same to every object but instance variables may vary to every new object

obj1 = A("Hey")
print("Data in obj1")
print(obj1.x)
print(obj1.y)

obj2 = A("Hello")
print("Data in obj2")
print(obj2.x)
print(obj2.y)

obj3 = A("Hi")
print("Data in obj3")
print(obj3.x)
print(obj3.y)
```

```python
# class variable modification
# syntax: 
# class_name.class_variable = new_value
A.x = 40
print(obj1.x)
print(obj2.x)
print(obj3.x)

# if we modify class variable using object it (only changed)creates new instance variable in that object not in whole class
# by this new variable is created in object, but x variable in class won't changes

obj1.x = 100
print(obj1.x)
print(obj2.x)
print(obj3.x)

# firstly it will search in instance/object variable (local variable/property) if it is not there in instance variable block then it will move to the class variable(global variable) if it is there in class variable it will return class variable value
```

```python
class A:
    x = 20
    def __init__(self,y):
        """hello py20""" #doc: documentation
        self.y = y
obj1 = A("hey")
print(A.__dict__)
print(obj1.__dict__)
```

```python
class A:
    x = 20
    def __init__(self,y):
        self.y = y
obj1 = A("hey")
obj2 = A("hello")
obj3 = A("hi")

print(A.__dict__)
print(obj1.__dict__)
print(obj2.__dict__)
print(obj3.__dict__)

# modify
obj1.x = 100 # these changes and updated new values are happens only in which object we are modifying but not in class variables or whole class or in other objects
obj1.z = 200
print(obj1.__dict__)
print(obj2.__dict__)
print(obj3.__dict__)

# object dictionary variables are instance varialbes and if we modify in class variable using object then new varialbes(instance variables)  are created
# by this class variables will not modified but new object/instance variables will be created, that new object will be shadowing.

obj2.__dict__['z'] = 40
print(obj2.__dict__)

# shadowing: shadowing done only to particular new instance variable that when we modify or update it 

# if we modify or update instance variables using objects then it will only changed in only that particular object but not changed in class or in other objects

# here we use __dict__ to know what variables are updated using object and where they are modifying
```
