```python
# customization
# Customization in OOP means modifying or designing classes, constructors, and object according to our specific requirements. We can define our owwn attributes, methods, constructor parameters, and object values.
# creating something according to our own requirement.

# Class
#  ↓
# We decide attributes + methods
#  ↓
# Constructor
#  ↓
# We decide what data to initialize
#  ↓
# Objects
#  ↓
# Each object can have different data

# ducktyping
# duck typing is a type system used in dynamic languages(dynamically typed). 
# For example Python,Perl,Ruby,PHP,Javascript,etc. where the type or the class of an object is less important than the method it defines. 
# Using Duck Typing, we do not check types at all. 
# Instead, we check for the presence of a given method or attribute.

# The name Duck Typing comes from the phrase:
# "If it looks like a duck and quacks(sounds) like a duck, it's a duck"

# Definition:
# Duck Typing means Python cares about what an object can do, rather than what type/class the object belongs to.
# This idea is:
# If it walks like a duck and quacks like a duck, then treat it like a duck.

# In python, if an object has required method or behavior, we can use it without checking its class.

class Dog:
    def speak(self):
        print("Dog says woof")
class Cat:
    def speak(self):
        print("Cat says Meow")
def make_sound(animal):
    animal.speak()

d = Dog()
c = Cat()
make_sound(d)
make_sound(c)
# If the object has a speak() method, it works.

# Duck typing is a python concept where the type of an object is less important than the methods or behavior it provies. 
# If an object supports the required operation, Python allows it to be used.
# Ducktyping = behavior matters more than type.
```
