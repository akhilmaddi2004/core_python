## Class Object Constructor

```python
# Everything in Python is an object to create objects of custom type we define a class.
# A class is a blueprint and object is the implementation or the actual thing built from 
class A:
    x = 20
obj1 = A()
obj2 = A()
print(type(obj1))
```

```python
# Key Terminology:
# i) class: A class is a blueprint
# ii) Object: The actual implementation of a class
# iii) Instance: Instance is another word for object
# iv) Attribute: Data stored inside an object
# v) Method: A function defined inside a class.
# vi) Class variable and instance/object variables: Python has two kinds of variables in a class

#       class variable                                          # Instance/object variables
# Defined directly inside the class body                # Defined inside __init__ using self.variable_name = value
# shared across all objects                             # Each object has its on copy
# Same value for every instance unless overridden       # Different objects can have different values.
# You can access it using :                             # You can access it using self or object
# className.variable = A.x/self.x                       # self.y/obj.y
# You can change the value using className.variable     # You can change the value using object.
```
