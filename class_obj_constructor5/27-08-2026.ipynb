```python
# A bank wants to create a simple system to store customer account details. 
# Create a class BankAccount with a class variable bank_name = "ABC Bank" and instance variables account_holder, account_number, and balance. 
# Initialize the instance variables using a constructor. 
# The constructor should validate that the initial balance is not negative; if it is negative, set the balance to 0. 
# Create two account objects and display their details.

class BankAccount:
    bank_name = "ABC Bank"
    def __init__(self,account_holder,account_number,balance):
        self.account_holder = account_holder
        self.account_number = account_number
        if(balance<0):
            self.balance = 0
        else:
            self.balance = balance
acc1 = BankAccount("Akhil",1234567890,-500)
acc2 = BankAccount("Abhi",7834299980,1500)
def dispaly(obj):
    print("Bank Account:",obj.bank_name)
    print("Account Holder:",obj.account_holder)
    print("Account Number:",obj.account_number)
    print("Balance:",obj.balance)
dispaly(acc1)
dispaly(acc2)

print(acc1.bank_name)
print(acc1.account_holder)
print(acc1.account_number)
print(acc1.balance)

print(acc1.__dict__)
print(acc2.__dict__)
```

```python
# A college wants to maintain student records. 
# Create a class Student with a class variable college = "ABC College" and instance variables name, roll_no, and marks. 
# Initialize these values using __init__(). 
# The constructor should validate that marks are between 0 and 100; if invalid marks are provided, set them to 0. 
# Create three student objects and display their details.

class Student:
    college = "ABC College"
    def __init__(self,name,rollno,marks):
        self.name = name
        self.rollno = rollno
        if(marks>=0 and marks<=100):
            self.marks = marks
        else:
            self.marks = 0
st1 = Student("Akhil",501,-90)
st2 = Student("Abhi",502,90)
st3 = Student("Arun",503,190)

def display(obj):
    print("College:",obj.college)
    print("Name:",obj.name)
    print("rollno:",obj.rollno)
    print("marks:",obj.marks)
display(st1)
display(st2)
display(st3)
print(st1.__dict__)
print(st2.__dict__)
print(st3.__dict__)
```

```python
# An online store wants to maintain a list of products added to its system. 
# Create a global list products = []. 
# Create a class Product with a class variable store_name = "ABC Store" and instance variables name, price, and quantity. 
# Initialize the values using the constructor and validate that price and quantity cannot be negative. 
# Whenever a product object is created, add its name to the global products list. 
# Create three products and display the product details and the complete product list.

products = []
class Product:
    store_name = "ABC store"
    def __init__(self,name,price,quantity):
        self.name = name
        if(price>0 and quantity>0):
            self.price = price
            self.quantity = quantity
        else:
            print("Invalid Price or Quantity")
        products.append(name)
prod1 = Product("Laptop",50000,2)
prod2 = Product("Mobile",30000,1)
prod3 = Product("Bags",500,2)
prod4 = Product("Clothes",500,-1)


def display(obj):
    print("Store Name:",obj.store_name)
    print("Product Name:",obj.name)
    print("Product Price:",obj.price)
    print("Product Quantity:",obj.quantity)
display(prod1)
display(prod2)
display(prod3)
print(prod1.__dict__)
print(prod2.__dict__)
print(prod3.__dict__)
print(prod4.__dict__)
print(products)
```
