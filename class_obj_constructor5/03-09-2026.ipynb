## isinstance

```python
# create a python application that takes 2 inputs from user which are age and name
# as instance variable for the class student.
# check whether the age is of integer type and name is string
# achieve this using 3 ways type, class, isinstance and do this validation process in the constructor.
# display error message if required
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        if(type(age)!=int):
            print("Error: Age must be integer")
        elif(type(name)!=str):
            print("Error: Name must be string")
        else:
            print("Valid Student")
name = input("Enter your name:")
age = int(input("enter your age:"))
s1 = Student(name,age)
```

```python
# by using type
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        if(type(age) != int):
            print("Error: Age must be integer")
        elif(type(name) != str):
            print("Error: Name must be string")
        else:
            print("Valid Student")
s1 = Student("Akhil",22)
s2 = Student(22,"Akhil")
s3 = Student(22,22)
```

```python
# using class
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        if(age.__class__!=int):
            print("Error: Age must be integer")
        elif(name.__class__!=str):
            print("Error: Name must be string")
        else:
            print("Valid Student")
s1 = Student("Akhil",22)
s2 = Student(22,22)
s3 = Student(22,"Akhil")
print(age.__class__)
print(name.__class__)
```

```python
# using isinstance
class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
        if(not isinstance(age,int)):
            print("Error: Age must be integer")
        elif(not isinstance(name,str)):
            print("Error: Name must be string")
        else:
            print("Valid Student")
s1 = Student("Akhil",22)
s2 = Student(22,22)
s3 = Student(22,"Akhil")
print(isinstance(age,int))
print(isinstance(name,str))
```

```python
# An online learning platform creates student enrollment objects. 
# Create a class Enrollment with class variable platform = "LearnOnline" and enrollment_count = 0. 
# The constructor should accept student_name, course, and age. 
# If the student’s age is 18 or above, set an instance variable status = "Registered"; otherwise set it to "Rejected". 
# Increase enrollment_count only for registered students. 
# Create three objects and use isinstance() to check which objects are instances of Enrollment.

class Enrollment:
    platform = "LearnOnline"
    enrollment_count = 0
    def __init__(self,student_name,course,age):
        self.student_name = student_name
        self.course = course
        self.age = age
        if(age>=18):
            self.status = "Registered"
            Enrollment.enrollment_count += 1
        else:
            self.status = "Rejected"
s1 = Enrollment("Akhil","Python",22)
s2 = Enrollment("Abhi","Java",23)
s3 = Enrollment("Mani","Python",17)
print(s1.status)
print(s2.status)
print(s3.status)
print("Registered Students:",Enrollment.enrollment_count)
print(isinstance(s1,Enrollment))
print(isinstance(s2,Enrollment))
print(isinstance(s3,Enrollment))
```
