## Error Handling

```python
# Python error handling relies on the try-except block to catch and manage runtime exceptions, preventing program crashes and allowing for graceful recovery.

# The structure includes "try" for risky code, "except" for specific error handling, "else" for success cases, and "finally" for mandatory cleanup tasks like closing flies.

# key mechanisms and best practices include:
# - Specifc Exception Handling: Catch specific exceptions(e.g., "ValueError", "ZeroDivisionError") rather than using bare "except" clauses to avoid masking unexcepted errors.

# - Raising Exceptions: Use the "raise"  keyword to trigger exceptions manually, enabling custom validataion logic(e.g., "raise ValueError("Invalid Input")").

# - Custom Exceptions: Developers can create custom exception classes by inherting from the base "Exception" class to provide domain-specific error messages.

# - Advanced Techniques: Modern Python (3.11+) supports exception groups and "except*" fro handling multiple errors simultaneously, while context managers("with" statements) are preferred for reliable resource management.

# - Best Practices: Follow the EAEP(Easier to Ask Forgiveness than Persmission) style, log errors appropriately, and let unknown exceptions propagate to ensure critical failures are not silently ignored.

try:
    result = 10/0
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Success:",result)
finally:
    print("Cleanup complete")
    # result = 10/0
```

```python
# Python Exception Handling
# Exception Handling allows a program to handle unexcepted errors during execution in a controlled way, instead of crashing abruptly.

# It enables programs to detect erros, manage them properly and continue execution wherever possible.

# - Handles runtime errors such as invalid input, file not found, division by zero and type mismatches that occur during program execution.
# - Helps improve program reliability by ensuring the application does not terminate unexpectedly when an error occurs.


# Without Exception Handling:
# If m is 0, the program will throw an exception(e.g., Division by Zero error) and crash.
n = 10
m = 0
# ans = n/m
# print(ans)
# ZeroDivisionError: division by zero

# With Exception Handling
# This code inclues exception handling to manage division by zero.
n = 10
m = 0
try:
    ans = n/m
    print(ans)
except ZeroDivisionError as zde:
    print("Error:Division by 0!")
    print(zde) # prints only error message

# keyword       What it does
# try block     # contains the code that might fail,like "ans= n/m".
# except        # Runs if a specific error (like "ZeroDivisionError") occurs in the "try" block.

n = 10
try:
    res = n/0
except ZeroDivisionError:
    print("Can't be divided by zero!")

# Explaination: Dividing a number by 0 raises a "ZeroDivisionError". The try block contains code that may fail and except block catches the error, printing a safe message instead of stopping the program.


# Syntax:
# Python provides four main keywords for handling exceptions: try, except, else, finally each plays a unique role: 
try:
    pass
except:
    pass
else:
    pass
finally:
    pass

# try: Runs the risky code that might cause an error.
# except: Catches and handels the error if one occurs.
# else: Executes only if no exception occurs in try.
# finally: runs regardless of what happens useful for cleanup tasks like closing files.

# Example: This code attempts division and handles error gracefully using try-except-else-finally.
try:
    n = 0
    res = 100/n
except ZeroDivisionError:
    print("You can't divide by zero!")
except ValueError:
    print("Enter a valid number!")
else:
    print("Result is",res)
finally:
    print("Execution complete.")

# Explanation: try block attempts division, except blocks catch specific errors, else block executes only if no errors occur, while finally block always runs, signaling end of the execution.


# Catching Exceptions.
# We can handle errors more efficiently by specifying the types of exceptions we except. This can make code both safer and easier to debug.

# 1. Specifc Exceptions:
# Catching specific exceptions make code to respond to different exception types differently.
# It precisely makes your code safer and easier to debug. It avoids masking  bugs by only reacting to the exact problems you except.

# EXample: This code handles ValueError and ZeroDivisionError with different messages.
print()
print('# 1. Specifc Exceptions:')
try:
    # This will cause ValueError
    x = int('str')
    inv = 1/x # Inverse calculation
except ValueError as ve:
    print("Not Valid!")
    print("ValueError:",ve)
except ZeroDivisionError as zde:
    print("Zero has no inverse!")
    print(zde)

# Explaination: A ValueError occurs because "str" cannot be converted to an integer. If conversion had succeeded but x were 0, a ZeroDivisionError would have been caught instead.

# 2. Multiple Exceptions:
print()
# We can catch multiple exceptions in a single block if we need to handle them in the same way or we can separate them if different types of exceptions require different handling.
# Example: This code attempts to convert list elements and handles ValueError, TypeError and IndexError.

a = ['10','twenty',30]
try:
    # 'twenty' cann't be convert to int
    total = int(a[0]) + int(a[1])
except (ValueError,TypeError) as e:
    print("Error:",e)
except IndexError:
    print("Index out of range.")

# Explaination: The ValueError is raised when trying to convert "twenty" to an integer. A TypeError could occur if incompatible types were used, while IndexError would trigger if the list index was out of range.

# 3. Catch-All Handlers and Their Risks
# Catch-All handler is used to call to catch any exception (similar to else statement). Use only except keyword to define it:
# Example: This code tries diving a string by a number, which causes a TypeError.
print()
try:
    # Risky operation: dividing string by number.
    res = "100"/20
except ArithmeticError as ae:
    print("ArithmeticError:",ae)
except Exception as e:
    print("Something went wrong!",e)

# Explaination:A TypeError occurs because you can’t divide a string by a number. The bare except catches it, but this can make debugging harder since the actual error type is hidden. Use bare except only as a net.

### Raise an Exception:
# We raise an exception using the "raise" keyword followed by an instance of the exception class that we want to trigger. We can choose from built-in exceptions or define our own custom exceptions by inheriting from Python's built-in Exception class.

# Syntax: raise ExceptionType("Error Message")

# Example: This code raises a ValueError if an invalid age is given.
print()
def set(age):
    if(age<0):
        raise ValueError("Age cann't be negative.")
    print(f"Age set to {age}")
try:
    set(-5)
except ValueError as e:
    print(e)

# ## Error vs Exception:
# Errors and exceptions are bothh issues in a program, but they differ in severity and handling.
# Error: Issues in the program logic such as SyntaxError, etc. It occurs at compile time.
# Exception: Problems that occur at runtime and can be managed using exception handling (e.g., invalid input, missing files).
```
