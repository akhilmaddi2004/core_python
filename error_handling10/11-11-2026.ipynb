```python
# • Write a function named find_length(obj) that uses a loop to calculate the length of the given object without using the built-in len() function. 
# The function should return the calculated length if the object is iterable. 
# If a non-iterable object such as an integer is passed, the function should raise and handle a TypeError, and print an appropriate error message explaining what happens when an integer is sent as input.

def find_length(obj):
    c = 0
    try:
        for i in obj:
            c += 1
        return c
    except TypeError as t:
        print("TypeError: Integer object is not iterable")
print(find_length('123'))
print(find_length([10,20,30,40]))
find_length(123)
```

```python
def find_length(obj):
    try:
        c = 0
        if isinstance(obj,(list,tuple,str,set)):
            for i in obj:
                c += 1
            return c
        elif isinstance(obj, dict):
            for k,v in obj.items():
                c += 1
            return c
        else:
            raise TypeError(f"Object type '{type(obj).__name__}' is not iterable.")
    except TypeError as e:
        print(e)
print("List:",find_length([12,2,3,4]))
print("Tuple:",find_length((1,2,3)))
print("Set:",find_length({1,2,3}))
print("String:",find_length("slkfjienc"))
print("Dict:",find_length({'A':1,'B':2}))
print(find_length(123))
```

```python
# • Create a class BankAccount with an attribute balance. 
# Implement a method withdraw(amount) that raises an exception if the withdrawal amount is greater than the available balance.
class BankAccount:
    def __init__(self,balance):
        self.balance = balance
    def withdraw(self,amount):
        if(amount > self.balance):
            raise ValueError("Insufficient balance")
        # else:
        self.balance -= amount
        print("Remaining balance:",self.balance)
b = BankAccount(1000)
# b.withdraw(2000)
b.withdraw(200)
print(b.balance)
```

```python
# • Create a class UserInput with a method get_integer(value). 
# Handle ValueError and TypeError using separate except blocks.
class UserInput:
    def get_integer(self,value):
        try:
            num = int(value)
            print(num)
        except ValueError as ve:
            print("ValueError: Invalid integer value")
        except TypeError as te:
            print("TypeError: Cannot convert this value to integer")
u = UserInput()
u.get_integer(9)
u.get_integer("abc")
u.get_integer(None)
u.get_integer([24])
```

```python
# • Create a base class Shape with a method area() that raises NotImplementedError. 
# Create a child class Rectangle that overrides and implements the area method.
class Shape:
    def area(self):
        raise NotImplementedError("Sub class must implement area()")
class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth
    def area(self):
        return self.length*self.breadth
    # pass
r = Rectangle(10,5)
r.area()

# if we won't create area() in rectangle it will raise an error of NIE
```

```python
# • Create a class Service with a method that calls another method which raises an exception. 
# Catch and handle the exception in the Service class.
class Service:
    def process(self):
        try:
            self.calculate()
        except ValueError:
            print("Exception handled in Service class")
    def calculate(self):
        raise ValueError("Something went wrong")
s = Service()
s.process()
```

```python
# • Create a class Transaction with a method process() that uses try, except, and finally blocks to ensure a cleanup message is always printed.
class Transaction:
    def process(self):
        try:
            print("Transaction processing")
            raise ValueError("Transaction failed")
        except ValueError:
            print("Exception handled: Transaction error")
        finally:
            print("Cleanup completed")
t = Transaction()
t.process()
```
