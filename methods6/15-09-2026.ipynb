```python
# 1. Instance Methods
# - Instance Methods are the most common type of methods they operate on specific object.
# - and have access to both class variables and object variables throw self.
# - An Instance Method is any method inside a class that takes 'self' as its first parameter.
# - Self gives the method access to specific object.
# - You can call it on object.
# ex: obj.m1()
class A:
    x = 20
    def __init__(self,y):
        self.y = y
    def display(self):
        print("Hi")
obj = A(120)
obj.display()
# A.display(1)
```

```python
# bank 
class Bank:
    def __init__(self,accno,pin,balance):
        self.accno = accno
        self.pin = pin
        self.balance = balance
    def deposit(self,amount):
        if(amount>0):
            self.balance += amount
            # print("Amount Deposited Successfully.")
            print(f"Your account XXXXX{self.accno%10000} has been credited with amount Rs. {amount}.")
            # print("Your balance:",self.balance)
        else:
            print("Invalid Amount")
    def withdraw(self,amount):
        if(amount>0 and amount<=self.balance):
            self.balance -= amount
            print(f"Your account XXXXX{self.accno%10000} has been debited with amount Rs. {amount}.")
            # print("Your balance:",self.balance)
        else:
            if(amount<=self.balance):
                print("Insufficient Funds.")
            else:
                print("Invalid Amount.")
    def check_balance(self):
        print("Your current balance:",self.balance)

acc1 = Bank(123400,4444,100000)
acc2 = Bank(567800,5555,1500)
acno = int(input("Enter Your account number:"))

if(acno == acc1.accno):
    pin = int(input("Enter your pin:"))
    if(pin == acc1.pin):
        print("Login Successful!")
        while(True):
            print("1. Deposit\n2. Withdraw\n3. Check Balance")
            choice = int(input("Enter your choice:"))
            if(choice == 1):
                amount = int(input("Enter the amount you want to deposit:"))
                acc1.deposit(amount)
            elif(choice==2):
                amount = int(input("Enter the amount you want to withdraw:"))
                acc1.withdraw(amount)
            elif(choice==3):
                acc1.check_balance()
            else:
                print("Invalid Choice")
                break
    else:
        print("Incorrect Pin")

elif(acno == acc2.accno):
    pin = int(input("Enter your pin:"))
    if(pin == acc2.pin):
        print("Login Successful!")
        while(True):
            print("1. Deposit\n2. Withdraw\n3. Check Balance")
            choice = int(input("Enter your choice:"))
            if(choice == 1):
                amount = int(input("Enter the amount you want to deposit:"))
                acc2.deposit(amount)
            elif(choice==2):
                amount = int(input("Enter the amount you want to withdraw:"))
                acc2.withdraw(amount)
            elif(choice==3):
                acc2.check_balance()
            else:
                print("Invalid Choice")
                break
    else:
        print("Incorrect Pin")

else:
    print("Invalid Acc_no")
```

```python
# 2. Class Methods
# - A Class Method is bound to a class rather than individual object.
# - It receives 'cls' as its first paramter rather than 'self'.
# - It is decorated with @classmethod above its method definition.
# - It can access and modify only class properties/variables not object properties.
# - It is called on object or class.
class A:
    x = 10
    y = 20
    # constructor
    def __init__(self,z):
        self.z = z
    # instance
    def update(self,a):
        self.a = a
    @classmethod
    def change(cls,new_value):
        cls.x = new_value
A.change(20)
print(A.x)

# obj = A(30)
# print(obj.x)
```

```python
# 3. Static Methods
# - A Static Method is a regular function that lives inside a class for organizational purposes.
# - It has no access to self or cls and it is completely independent of both.
# - It is decorated with @staticmethod above the method definition no automatic first argument.
# - It can called on an object and class.
class A:
    x = 10 
    y = 20
    def __init__(self,z):
        self.z = z
    @staticmethod
    def check_age(age):
        return age>18
    @classmethod
    def change(cls,new_value):
        cls.x = new_value
print(A.check_age(20))
obj = A(30)
print(obj.check_age(17))

A.change(30)
print(A.x)

# the methods which are not depend on class or obj property/methods they are static methods and these are independent.
# without @classmethod or @staticmethod class takes by default any method as instance method.
# static method is same for both class & obj.
# If we call it with class or obj we get same output by static method
```

```python
# 1. Create a class Student that stores a student’s name, roll number, and marks. 
# Create an instance method to display the student’s details and another instance method to calculate the student’s percentage. 
# Create a class method to keep track of the total number of students created. 
# Also create a static method that accepts marks and returns "Pass" if the marks are 40 or above, otherwise "Fail". 
# Create multiple student objects and test all the methods.
class Student:
    no_of_students = 0
    def __init__(self,name,rollno,marks):
        self.name = name
        self.rollno = rollno
        self.marks = marks
        Student.no_of_students += 1
    def display(self):
        print("Name:",self.name)
        print("rollno:",self.rollno)
        print("Marks:",self.marks)
    def percentage(self,max_marks = 100):
        percent = (self.marks/max_marks)*100
        return percent
    @classmethod
    def track(cls):
        return cls.no_of_students
    @staticmethod
    def grade(marks):
        if(marks>=40):
            return "Pass"
        else:
            return "Fail"

s1 = Student("Akhil",101,85)
s2 = Student("Rahul",102,35)
s3 = Student("Priya",103,72)

s1.display()
print("Percentage:",s1.percentage())
print("Result:",Student.grade(s1.marks))
print()

s2.display()
print("Percentage:",s2.percentage())
print("Result:",Student.grade(s2.marks))
print()

s3.display()
print("Percentage:",s3.percentage())
print("Result:",Student.grade(s3.marks))
print()

print("Total Students:",Student.track())
```

```python
# 4. Write a class PasswordUtils with three @staticmethod methods:
# is_strong(password) - returns True if length>=8, has a digit, and has an uppercase;
# generate_hint(password) - returns first 2 chars +'***';
# validate_email(email) - returns Ture if '@' and "." are present.
# Test all three.
class PasswordUtils:
    @staticmethod
    def is_strong(password):
        has_digit = False
        has_uppercase = False
        for i in password:
            if(i.isdigit()):
                has_digit = True
            if(i.isupper()):
                has_uppercase = True
        if(len(password) >= 8 and has_digit and has_uppercase):
            return True
        else:
            return False
    @staticmethod
    def generate_hint(password):
        return password[:2]+"*****"
    @staticmethod
    def validate_email(email):
        if("@" in email and "." in email):
            return True
        else:
            return False
print(PasswordUtils.is_strong("Akhil123"))
print(PasswordUtils.is_strong("akhil"))
print(PasswordUtils.is_strong("1244"))
print()
print(PasswordUtils.generate_hint("Akhil1234"))
print()
print(PasswordUtils.validate_email("akhilmaddi@gamil.com"))
print(PasswordUtils.validate_email("akhilmaddigamil.com"))
print(PasswordUtils.validate_email("akhilmaddigamilcom"))
```
