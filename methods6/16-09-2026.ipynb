```python
# 2. ⁠Create a class BankAccount that stores account holder name, account number, and balance. 
# Create instance methods to deposit and withdraw money and to display the current balance. 
# Create a class method to display the bank name, which is common to all accounts. 
# Create a static method to check whether a withdrawal amount is valid, where the amount must be greater than zero. 
# Create two account objects and test the methods.

class Bank:
    bank_name = "SBI"
    def __init__(self,account_holder,accno,balance):
        self.account_holder = account_holder
        self.accno = accno
        self.balance = balance
    def deposit(self,amount):
        self.balance += amount
        print("Your account is credited with Rs.",amount,"successfully")
    def withdraw(self,amount):
        if(self.balance<=amount):
            if(self.check(amount)):
                self.balance -= amount
                print("Your account is debited with Rs.",amount,"Successfully")
            else:
                print("Invalid Amount")
        else:
            print("Insufficient Funds")
    def check_balance(self):
        print("Your available balance is",self.balance)
    @classmethod
    def display_bank_name(cls):
        print("Your bank name is",cls.bank_name)
    @staticmethod
    def check(amount):
        return amount>0
b1 = Bank("A1",1234,10000)
b2 = Bank("A2",3456,15000)
b3 = Bank("A3",6789,100)
Bank.display_bank_name()
b1.display_bank_name()
b1.deposit(1200)
b1.withdraw(1000)
b1.check_balance()
print()
b2.display_bank_name()
b2.deposit(1200)
b2.withdraw(1000)
b2.check_balance()
print()
b3.display_bank_name()
b3.deposit(1200)
b3.withdraw(1000)
b3.check_balance()

print(Bank.check(0))
print(Bank.check(-100))
print(Bank.check(100))
```
