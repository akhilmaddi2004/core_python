```python
# 2. Create a function student_info(name, *subjects, **details) that prints a student's name,subjects enrolled, and additional details like grade and school.
def student_info(name,*subjects,**details):
    print("\t \t \t \tStudent Info:")
    print("Student Name:",name)
    print("Subjects enrolled by student:")
    for subject in subjects:
        print(subject,end=",")
    print("\nAdditional Details of student:")
    for detail,about in details.items():
        print(f"{detail}:{about}")
    
student_info("Akhil",
             "Mathematics","Science","Social","English",
             grade=9,school="Sri Vani High School",age=15)
```

```python
# 3. Write a function order_food(*items,**perferences) that accepts multiple food items and optional perferences like spice level or delivery time.Display the order summary.
def order_food(*items,**perferences):
    print("Order Summary:")
    print("\nList of items:")
    for item in items:
        print(item)
    print("\nPerferences:")
    for perference,value in perferences.items():
        print(f"{perference}:{value}")
    
order_food("Biryani","Butterscotch IceCream","Coke","onions","lemons",
           spice_level = "medium",delivery_time="25 mins")
```

```python
# 4. Write a function shopping_cart(discount = 0, *prices) that calculates the total price after applying a discount. Demostrate calling the function with and without the discount argument.
def shopping_cart(discount = 0, *prices):
    total = 0
    for price in prices:
        total += price
    print("Total price before discount:",total)
    total_price = total - discount
    print("Total price after discount:",total_price)
shopping_cart(0,150,20,50,30)
shopping_cart(150,20,50,30)
# if didn't mention first positinoal argument for default discount parameter it will the first argument to the discount and next rest of them will be passed to the prices
# and discount will be substracted with total_price

# shopping_cart(150,20,50,30,discount=60)  

# shopping_cart(discount=0,150,20,50,30)

shopping_cart(50,150,20,50,30)
```

```python
# 5. Design a function register_user(username,role="user",*permisions,**details) that stores user information, including optional permissions and additional attributes.

def register_user(username,role="user",*permissions,**details):
    print("Username: ",username)
    print("Role: ",role)
    print("\nPermissions: ")
    if permissions:
        for permission in permissions:
            print("-",permission)
    else:
        print("No Permissions assigned")
    print("\nAdditional details:")
    if details:
        for detail,about in details.items():
            print(f"{detail}:{about}")
    else:
        print("No additional details")
register_user("Akhil",
              "admin",
              "create_user","delete_user",
              email = "akhil@gmail.com",
              age = 35)
print()
register_user("suresh",
              "developer",
              "commit","pull","push",
              email = "suresh@gmail.com",
              age = 24)
print()
register_user("rahul")
```

```python
# 7. Define a function login(username, password = "1234").Demonstrate how default arguments work and explain a potential issue with using default passwords.
def login(username, password = "1234"):
    print("User Login Credentials:")
    print("Username:",username)
    print("Password:",password)
login("Akhil")
login("Akhil","Akhilmaddi1@a")

# Default arguments allow callers to omit specific parameters. If the user provides only the username, Python automatically uses the default password. 

# With Default: login("alice") uses "1234" for the password.
# Overriding Default: login("bob", "secret") uses "secret" for the password.
# Potential Issue with Default Passwords

# Using a static default password like "1234" is a significant security risk. It allows unauthorized access if the caller fails to provide a secure password. Additionally, hardcoding credentials makes the system vulnerable to basic attacks and violates best practices for authentication, which typically require checking against a secure database of hashed passwords rather than using plaintext defaults.
```

```python
# 8. Write a function area(length,breadth = None) that calculates the area of a rectangle. If breadth is not provided, assume it is a square and compute accordingly.
def area(length,breadth = None):
    if (breadth == None):
        # square
        print(length*length)
    else:
        # rectangle
        print(length*breadth)
area(4) # square
area(6,4) # rectangle
```

```python
# 9. Write a function calculate_score(base_score = 0, *bonus_points, **penalties) that computes a final score after adding bonuses and subtracting penalties.
def calculate_score(base_score = 0, *bonus_points, **penalties):
    
    add_bonus = 0
    for bonus in bonus_points:
        add_bonus += bonus
    final_score = base_score + add_bonus # adding bonus to base_score
    
    total_penalty = 0
    for penalty, value in penalties.items():
        # print(f"{penalty}:{value}")
        total_penalty += value
    final_score = final_score - total_penalty # subtracting penalty after adding bonus to base_score
    
    # total_score -= sum(penalties.values())
    
    print("Final score:",final_score)

calculate_score(0,
                1,2,5,1,
                penalty1 = 1, penalty2 = 2, penalty3 = 1)
calculate_score(100,
                10,5,15,
                late_submission = 5,
                absence = 10)
```

```python
# 10. Design a function send_email(sender, receiver, subject= "No Subject", *attachments,**options) that simulates sending an email with optional attachments and settings.

def send_email(sender,receiver,subject = "No Subject",*attachments, **options):
    print("Email details:")
    print("Sender email:",sender)
    print("Receiver email:",receiver)
    print("Subject:",subject)
    print("Attachments:")
    for attachment in attachments:
        print("-",attachment)
    print("Options:")
    for option,detail in options.items():
        print(f"-{option}:{detail}")

send_email("akhilmaddi@gmail.com",
           "chennuriabhilash@startup.in",
           "Applying for Developer role",
           "passport photo","aadhar card","resume",
           priority = "High",
           read_receipt = True)

print()

send_email("akhil@gmail.com",
           "xyz@gmail.com",
           "Project Update",
           "report.pdf",
           "presentation.pptx",
           priority = "High",
           read_receipt = True)
```
