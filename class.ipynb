```python
# 1. Create a function start_system()
#     Write a decorator that prints:
#     * “System starting…” before execution
#     * “System started successfully” after execution
def start_system():
    print("Starting System..")
def dec1(func):
    def wrapper1():
        func()
        print("System Started successfully")
    return wrapper1
start_system = dec1(start_system) # start_system will be updated with decorator #eg: x = 5, x=x+5
start_system() # calls wrapper1 
```

## short hand format decorators

```python
# implementing two-ways of decorators
# manual decorators
def decorator1(func):
    def wrapper(*args,**kwargs):
        print("Before calling")
        func(*args,**kwargs)
        print("After calling")
    return wrapper

def add(a,b):
    print(a+b)

add = decorator1(add)
add(20,30)
```

```python
# short-hand format
def decorator2(func):
    def wrapper(*args,**kwargs):
        print("Before calling")
        func(*args,**kwargs)
        print("After calling")
    return wrapper

@decorator2
def add(a,b):
    print(a+b)

add(20,30)
```

```python
def shout(func):
    def wrapper(*args,**kwargs):
        result = func(*args,**kwargs)
        x = result.upper()
        return x
        # return str(result).upper()
    return wrapper
@shout
def greet(name):
    return f"Hello! {name}"

print(greet("Akhil"))
```

```python
# 1.     Create a function get_message() that returns "hello user". Write a decorator using @ syntax that converts the output to uppercase.
def dec(func):
    def wrapper(*args,**kwargs):
        result = func(*args,**kwargs)
        # x = result.upper()
        # return x
        # return result.upper()
        return str(result).upper()
    return wrapper
@dec
def get_message():
    return "hello user"
print(get_message())
```

```python
# 2.     Create a function get_number() that returns 10
#     Use a decorator to return double the value.
def double(func):
    def wrapper(*args,**kwargs):
        res = func(*args,**kwargs)
        return res*2
    return wrapper
@double
def get_number(n):
    return n
print(get_number(10))
```
