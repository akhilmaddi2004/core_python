```python
# 1. Given a list of integers, write a program to filter even numbers and then multiply each of them by 2
# using a single pipeline of filter() and write your own version of map() called my_map(func,lst) using a regular loop.

lst = [1,2,3,4,5,6,7,8]
print(list(map(lambda x:x*2, list(filter(lambda x:x%2 == 0, lst)))))

def my_map(func,lst):
    l = []
    for i in lst:
        x = func(i)
        l.append(x)
    return l
def mul(x):
    return x*2
# lst1 = [1,2,3,4,5,6,7,8]
print(my_map(mul,lst))
print(my_map(mul,list(filter(lambda x:x%2 == 0 , lst))))
```

```python
# 2.Write a function greet(name, prefix = 'Hello', formatter = lambda x:x) that returns a greeting message.
# Apply the formatter functino to modify the output. Demonstrate the function using str.upper.
def greet(name,prefix='Hello',formatter = lambda x:x):
    x = f"{prefix},{name}!"
    return formatter(x)
print(greet("Akhil"))
print(greet("Akhil",formatter=str.upper))
print(greet("Akhil",formatter=lambda x:x.upper()))
print(greet("Akhil","Hi",formatter=str.upper))
```

```python
# 3. Given a list of tuples (name, marks), write a program to sort the list by marks in descending order.
lst = [
    ("Akhil",90),("Arjun",67),("Sara",78),("Gill",70),("Dhoni",59)
]
print(sorted(lst,key=lambda x:x[1],reverse=True))
```

```python
# 4. Given a list of product prices, write a program to filter prices greater than Rs. 500,
# apply a 10% discount using map(), and find the total price using reduce().
from functools import reduce
prices = [501,499,599,450,890,690]
prices = list(filter(lambda x:x>500, prices))
prices = list(map(lambda x:x-(x*0.1),prices))
prices = reduce(lambda x,y: x+y,prices)
print(prices)

prices1 = [501,499,599,450,890,690]
print(reduce(lambda x,y:x+y,list(map(lambda x: x-(x*0.1), list(filter(lambda x:x>500, prices1))))))
```

```python
# 5. Build a mini data pipeline.
# Start with a list of student dictionaries [{name,score}].
# Use filter() to keep scores>=60,
# map() to add a 'grade' key ('Pass'), and
# sorted() to sort by score descending.
# print the final result.
students= [
    {'name':'Akhil','score':78},
    {'name':'Mani','score':89},
    {'name':'Mahesh','score':56},
    {'name':'Raju','score':68},
    {'name':'Abhi','score':90},
]
results = list(filter(lambda d: d['score']>60, students))
graded = list(map(lambda x: {**x, 'grade':"pass"}, results))
# graded = list(map(lambda x:dict(x,grade = "pass"), results))
sorting = sorted(graded,key=lambda x:x['score'],reverse=True)
print(sorting)
```
