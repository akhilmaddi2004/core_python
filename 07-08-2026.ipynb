### Decorators with *args and **kwargs


```python
# The wrapper above onl works for functions with no arguments. To make a decorator work for ANY function, always use *args and **kwargs in the wrapper.
```

```python
# Universal wrapper template - ALWAYS use this pattern:
def my_decorator(func):
    def wrapper(*args,**kwargs):
        print("before")
        func(*args,**kwargs)
        print("After")
    return wrapper

def add(a,b):
    print(a+b)

add = my_decorator(add)
add(3,4)
add(3,b=4)
```

```python
def my_dec(func):
    def wrapper(*args,**kwargs):
        print("Before")
        result = func(*args,**kwargs) # psss all args through
        print("After")
        return result # pass return value through
    return wrapper
def add(a,b):
    return a+b

add = my_dec(add)
print(add(10,20))
```

```python
# known parameters
def add(a,b):
    print(a+b)

def dec1(func):
    def wrapper():
        print("before addition")
        func(10,20)
        print("After addition")
    return wrapper

add = dec1(add)
add()
```

```python
# unknown no. of parameters
def add(a,b):
    print(a+b)

def dec(func):
    def wrapper(*args,**kwargs):
        print("Before calling")
        func(*args,**kwargs)
        print("After calling")
    return wrapper

add = dec(add)
add(10,20)
add(10,b=20)
```

```python
def add(*args):
    sum = 0
    for i in args:
        sum += i
    print(sum)
def dec(func):
    def wrapper(*args,**kwargs):
        print("before")
        func(*args,**kwargs)
        print("after")
    return wrapper

add = dec(add)
add(1,34,22,34,52,12,34,90)
```

```python
def info(**kwargs):
    for k,v in kwargs.items():
        print(k,":",v)

def dec(func):
    def wrapper(*args,**kwargs):
        print("About sunil:")
        func(*args,**kwargs)
    return wrapper

info = dec(info)
info(name="sunil",age=23,city='srcl')
```

```python
# 1. Create a function start_system()
#     Write a decorator that prints:
#     * “System starting…” before execution
#     * “System started successfully” after execution
def start_system():
    print("System starting...")

def dec(func):
    def wrapper():
        func()
        print("System started successfully")
    return wrapper

start_system = dec(start_system) # start_system will be updated with decorator #eg: x = 5, x=x+5
start_system() # calls wrapper1 
```

```python
# 2.     Create a function show_message()
#     Write a decorator that prints:
#     * “Welcome!” before
#     * “Goodbye!” after
def show_message():
    print("Showing the message")
def dec2(func):
    def wrapper2():
        print("Welcome!")
        func()
        print("Done displaying the message.")
    return wrapper2
# x = dec2(show_message()) #TypeError: 'NoneType' object is not callable
# x() #TypeError: 'NoneType' object is not callable

show_message = dec2(show_message)
show_message()
```
