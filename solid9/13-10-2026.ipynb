```python
from abc import ABC,abstractmethod
class Netflix(ABC):
    def __init__(self,username,pwd,device):
        self.username = username
        self.pwd = pwd
        self.save_device = device
class ContentManager(Netflix):
    @abstractmethod
    def play(self):
        pass
class Downloadable(Netflix):
    @abstractmethod
    def download(self):
        pass
class Movies(ContentManager,Downloadable):
    def __init__(self, username, pwd, device):
        super().__init__(username, pwd, device)
    def play(self):
        print("Playing a movie!")
    def download(self):
        print(f"Downloading the movie to {self.save_device.__class__.__name__}")
class Series(ContentManager,Downloadable):
    def __init__(self, username, pwd, device):
        super().__init__(username, pwd, device)
    def play(self):
        print("Playing a Series!!")
    def upload(self):
        print("Uploading a new Season!!")
    def next(self):
        print("Playing next episode.")
    def prev(self):
        print("Previous Episode.")
    def download(self):
        print(f"Downloading the series!! {self.save_device.__class__.__name__}")
        print("Previous Episode.")
class Documentary(ContentManager,Downloadable):
    def __init__(self, username, pwd, device):
        super().__init__(username, pwd, device)
    def play(self):
        print("Playing a Documentary!")
    def download(self):
        print(f"Downloading the documentary!! {self.save_device.__class__.__name__}")
class Trailer(ContentManager):
    def play(self):
        print("Playing a Trailer!")
class PaymentManager(Netflix):
    @abstractmethod
    def pay(self):
        pass
class NotificationManager(Netflix):
    @abstractmethod
    def notify(self):
        pass
class Device(ABC):
    @abstractmethod
    def save(self):
        pass
class Phone(Device):
    def save(self):
        print("Saving data to your phone.")
class Tablet(Device):
    def save(self):
        print("Saving data to your tablet.")
class Desktop(Device):
    def save(self):
        print("Saving data to your desktop.")
class TV(Device):
    def save(self):
        print("Saving data to your tv.")
def play_content(content):
    content.play()
movie = Movies("alice123",123,TV())
play_content(movie)
play_content(Series("alice123",123,Desktop()))
play_content(Documentary("alice123",123,Desktop()))
play_content(Trailer("alice123",123,Desktop()))
movie.download()
```

```python
# Principle	Simple meaning
# S — Single Responsibility     A class should have one main responsibility.
# O — Open/Closed	            Add new features without changing existing working code.
# L — Liskov Substitution	    A child object should work wherever its parent type is expected.
# I — Interface Segregation	    Use small, specific interfaces instead of forcing classes to implement unnecessary methods.
# D — Dependency Inversion	    High-level code should depend on abstractions, not concrete classes.

# 1. S -> Single Responsibility Principle(SRP):
# One class should have one main responsibility
# In our code, Netflix stores user information, Device handles saving, content classes handle playing and downloading.

from abc import ABC, abstractmethod

# Responsibility : store user information
class Netflix:
    def __init__(self,username,pwd,device):
        self.username = username
        self.pwd = pwd
        self.device = device
# Responsibility : Play content
class ContentManager(ABC):
    @abstractmethod
    def play(self):
        pass
# Responsibility : Download Content
class Downloadable(ABC):
    @abstractmethod
    def download(self):
        pass
# Reponsibility : Play and download movies
class Movies(ContentManager,Downloadable):
    def __init__(self,username,pwd,device):
        self.user = Netflix(username,pwd,device)
    def play(self):
        print("Playing a movie!")
    def download(self):
        print("Downloading the movie!")
# Responsibility : Play and download series
class Series(ContentManager,Downloadable):
    def play(self):
        print("Playing a Series!")
    def download(self):
        print("Downloading the series!")
# Responsiblity : Save data on a device
class Device(ABC):
    @abstractmethod
    def save(self):
        pass
class Phone(Device):
    def save(self):
        print("Saving Data to phone.")
m = Movies("Alice123",123,Phone())
m.play()
m.download()

# - Netflix stores user information
# - ContentManager defines the playing contract.
# - Downloadable defines the downloading contract.
# - Device defines the saving contract.
# - Movies handles specific behaviour.
# Each class has a clear responsibility
```

```python
# 2. O - Open/Closed Principle(OCP):
# Meaning: A class or function should be open for extension but closed for modification.
# Your original play_content(content) function is a good example. We can add a new content type without changing that function.
from abc import ABC,abstractmethod
class ContentManager(ABC):
    @abstractmethod
    def play(self):
        pass
class Movies(ContentManager):
    def play(self):
        print("Playing a movie!")
class Series(ContentManager):
    def play(self):
        print("Playing a Series!")
class Documentary(ContentManager):
    def play(self):
        print("Playing a Documentary!")
class Trailer(ContentManager):
    def play(self):
        print("Playing a Trailer!")
# Exisiting function
def play_content(content):
    content.play()
# Exisiting content
play_content(Movies())
play_content(Series())
# New content can be added without modifying play_content()
class ShortFilm(ContentManager):
    def play(self):
        print("Playing a Short Film!")
play_content(ShortFilm())

# How OCP is applied:
# - When you introduce ShortFilm, you don't need to modify play_content().
# - You only create a new class implementing play().
# - Remember: Add new classes instead of repeatedly changing existing working code.
```

```python
# 3. L - Liskov Substitution Principle(LSP):
# Meaning: A child object should be usable wherever its parent type is expected, without breaking the program's expected behavior.
# In your example, Movies, Series, Documentary, and Trailer all inherit from ContentManager and implement play().

from abc import ABC, abstractmethod
class ContentManager(ABC):
    @abstractmethod
    def play(self):
        pass
class Movies(ContentManager):
    def play(self):
        print("Playing a movie!")
class Series(ContentManager):
    def play(self):
        print("Playing a Series!")
class Documentary(ContentManager):
    def play(self):
        print("Playing a Documentary!")
class Trailer(ContentManager):
    def play(self):
        print("Playing a Trailer!")
def play_content(content):
    content.play()
contents = [Movies(),Series(),Documentary(),Trailer()]
for content in contents:
    play_content(content)

# How LSP is applied:
# - play_content() expects an object that follows the ContentManager contract. Each child can be passed to it and works correctly.
# - Notice that Trailer does not implement download().
# - That is fine because this function only requires play().
# - Remember: Child classes should honor the behavior promised by their parent abstraction.
```

```python
# 4. I -> Interface Segregation Principle(ISP):
# Meaning: A class should not be forced to implement methods it does not need.
# Your code already demonstrates this well because you separated ContentManger and Downloadable.
from abc import ABC,abstractmethod
class ContentManager(ABC):
    @abstractmethod
    def play(self):
        pass
class Downloadable(ABC):
    @abstractmethod
    def download(self):
        pass
class Movies(ContentManager,Downloadable):
    def play(self):
        print("Playing a movie!")
    def download(self):
        print("Downloading the movie!")
class Series(ContentManager,Downloadable):
    def play(self):
        print("Playing a Series!")
    def download(self):
        print("Downloading the Series!")
# Trailer only needs to play content
class Trailer(ContentManager):
    def play(self):
        print("Playing a Trailer!")
m = Movies()
m.play()
m.download()
t = Trailer()
t.play()
# How ISP is applied:
# - Movies needs both playing and downloading.
# - Series needs both playing and downloading.
# - Trailer only needs playing.
# If play() and download() were compulsory methods in one large abstract class, Trailer would be forced to implement a download method it doesn't need.
# By separating the interfaces, each class implements only the methods it requires.
# Remember: Small, specific interfaces are better than one large interface that forces unnecessary methods.
```

```python
# 5. D - Dependency Inversion Principle(DIP):
# Meaning: High-level code should depend on abstractions rather than specific implementations.
# In your code, Movies needs a device to save downloaded content. Instead of depending directly on Phone,TV,or Desktop, it can depend on the abstract Device class.

from abc import ABC,abstractmethod
class Device(ABC):
    @abstractmethod
    def save(self):
        pass
class Phone(Device):
    def save(self):
        print("Saving data to your phone.")
class Desktop(Device):
    def save(self):
        print("Saving data to your desktop.")
class TV(Device):
    def save(self):
        print("Saving data to your TV.")

class Downloadable(ABC):
    @abstractmethod
    def download(self):
        pass
class Movies(Downloadable):
    def __init__(self,device):
        self.save_device = device
    def download(self):
        print("Downloading the movie!")
        self.save_device.save()
# The movie uses the Device abstraction
m1 = Movies(Phone())
m1.download()
print()
m2 = Movies(Desktop())
m2.download()
print()
m3 = Movies(TV())
m3.download()

# How DIP is applied :
# Movies does not need to know the internal details of each device. It only expects a device that implements save().
# You can pass a Phone,Desktop, or TV without changing the Movies class
# Remember: Depend on the abstract Device contract, not on a particular device class.
```
