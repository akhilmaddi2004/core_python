## SOLID Principles

```python
# The SOLID Principles are five object-oriented design guidelines introduced by RobertC. Martin to create flexible, maintainable, and scalable software.

# In Python, these principles help developers write clean code that is easy to test, extend, and reason about.

# The five principles are:
# 1. Single Responsibility Principle(SRP): A class should have only one reason to change, meaning it should handle a single job or functionality.

# 2. Open/Closed Principle(OCP): Software entities should be open for extension but closed for modification, allowing new features without altering existing code.

# 3. Liskov Substitution Principle (LSP): Objects of a superclass should be replaceable with objects of its subclasses without breaking the program's correctness

# 4. Interface Segregation Principle (ISP): Clients should not be forced to depend on interfaces or methods they do not use.

# 5. Dependency Inversion Principle (DIP): High-level modules should not depend on low-level modules, both should depend on abstractions.
```
