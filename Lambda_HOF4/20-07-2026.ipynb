```python
# Given a list of product prices, write a program to:
# Filter prices greater than 500
# Apply a 10% discount to the filtered prices using map()
prices = [500,200,300,1000,700]
prices = list(filter(lambda x: x>500, prices))
prices = list(map(lambda x: x-(x*0.1) , prices))
print(prices)
```

```python
# 1. Given a list of integers, write a program to filter even numbers and then multiply each of them by 3 using a single pipeline.
integers = [10,9,18,17,90,99,107]
integers = list(filter(lambda x: x%2 ==0,integers))
integers = list(map(lambda x:x*3, integers))
print(integers)

print("Single pipe line")
integers1 = [10,9,18,17,90,99,107]
print(list(map(lambda x:x*3, list(filter(lambda x: x%2==0, integers1)))))
```

# Reduce() 

```python
# reduce() function -> reduce function returns single value from given iterable
from functools import reduce
l = [1,2,3,4,7,9,14]
print(reduce(lambda x,y:x+y,l))
```

```python
# max value x>y             min value x<y
l = [1,17,28,4,14,140]
print("max value",reduce(lambda x,y: x if x>y else y, l)) # max in iterables

print("max value",reduce(lambda x,y:max(x,y), l)) # max in iterables

print("min value",reduce(lambda x,y: x if x<y else y, l)) #min value
```

```python
# Reduce: reduce()
# syntax: reduce(function,iterable)
# Reduce applies a function cumulatively the result of each step is pass to the next as argument(first argument)
# we can do operations like product, sum, avg, min, max of iterables
```

```python
#  concatenation of list of strings using results
strings = ['hi','hello','bye']
print(reduce(lambda x,y: x+y,strings))
print(reduce(lambda x,y: x+" "+y, strings))
print(reduce(lambda x,y: x+"@"+y, strings))
```

```python
# 1. Given a list of strings, write a program using reduce() to concatenate all strings into a single string.
lst = ['vamshi','abhi','akhil']
print(reduce(lambda x,y: x+y,lst))
```

```python
# 2. Given a list of digits, write a program using reduce() to form a single number (e.g., [1,2,3] → 123).
digits = [1,2,3]
print(reduce(lambda x,y:str(x)+str(y), digits))

print(reduce(lambda x,y: x*10+y, digits))

# For [1, 2, 3]:

# Start with 1
# 1 * 10 + 2 = 12
# 12 * 10 + 3 = 123

# So the final result is 123.
```

```python
# 3. ⁠Given a list of numbers, write a program using reduce() to calculate the cumulative difference
nums = [90,93,24,78,67]
print(reduce(lambda x,y: x-y,nums))
```
