```python
# 1. Given a list of product prices, write a program to filter prices above ₹500, then apply a 10% discount using map(),
# and compute the final total bill using reduce().
from functools import reduce
lst = [501,500,499,299,699,169,799]

lst = list(filter(lambda x: x>500, lst))
print(lst)
lst = list(map(lambda x: x-(x*0.1),lst))
print(lst)
lst = reduce(lambda x,y:x+y,lst)
print(lst)
```

```python
prices = [501,500,499,299,699,169,799]
print(reduce(lambda x,y: x+y,list(map(lambda x: x-(x*0.1),list(filter(lambda x: x>500, prices))))))
```

```python
# 2. Given a list of numbers, write a program to filter negative numbers,
# then convert them into positive numbers using map(), and find their sum using reduce().

numbers = [2,-5,9,-6,4,-8]
numbers = list(filter(lambda x:x<0, numbers))
print(numbers)
numbers = list(map(lambda x: abs(x), numbers))
print(numbers)
numbers = reduce(lambda x,y: x+y,numbers)
print(numbers)
```

```python
number = [2,-5,9,-6,4,-8]
print(reduce(lambda x,y: x+y, list(map(lambda x:x*(-1), list(filter(lambda x:x<0, number))))))
```

```python
# 3. Given a list of integers, write a program to filter numbers less than 50, then multiply each by 3 using map(),
# and determine the maximum value using reduce().
integers = [23,55,34,75,18,90]
integers = list(filter(lambda x: x<50, integers))
print(integers)
integers = list(map(lambda x:x*3, integers))
print(integers)
integers = reduce(lambda x,y: x if x>y else y,integers)
print(integers)
```

```python
integer = [23,55,34,75,18,90]
print(reduce(lambda x,y : x if x>y else y, list(map(lambda x:x*3, list(filter(lambda x:x>50, integer))))))
```

```python
# 4. Given a list of words, write a program to filter words with length greater than 3,
# then convert them to uppercase using map(), and concatenate them into a single string using reduce().

words = ['akhil','abhi','sunil','vamshi','anu']
words = list(filter(lambda x: len(x)>3, words))
print(words)
words = list(map(lambda x: x.upper(), words))
print(words)
words = reduce(lambda x,y:x+y,words)
print(words)
```

```python
word = ['akhil','abhi','sunil','vamshi','anu']
print(reduce(lambda x,y:x+y, list(map(lambda x: x.upper(),list(filter(lambda x: len(x)>3, word))))))
```

```python
# 5. A company tracks employee salaries. Write a program to filter salaries greater than ₹30,000,
# increase them by 15% using map(), and compute the total salary expenditure using reduce().
l3 = [20000,30000,35000,40000,29000,42000]
l3 = list(filter(lambda x:x>30000, l3))
print(l3)
l3 = list(map(lambda x:x+(x*0.15),l3))
print(l3)
l3 = reduce(lambda x,y:x+y,l3)
print(l3)
```

```python
# 8. A banking system stores transaction amounts. Write a program to filter only credit transactions (positive values),
# apply a processing bonus of ₹10 to each using map(),
# and calculate the total credited amount using reduce().

l6 = [-20000,-6000,10000,17000]
l6 = list(filter(lambda x: x>0, l6))
print(l6)
l6 = list(map(lambda x:x+10, l6))
print(l6)
l6 = reduce(lambda x,y: x+y, l6)
print(l6)
```
