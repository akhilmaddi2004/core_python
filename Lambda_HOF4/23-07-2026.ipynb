# Recursion

```python
# Recursion is a programming technique where a function calls itself either directly or indirectly to solve a problem. It is commonly used for:
# 1. Breaking problems in smaller subproblems
# 2. Mathematical calculations like factorial and Fibonacci
```

```python
# In recursion we can move forward (means recursion) and move to backwoard (back-tracking)
# Recursion works same as loops but we can't move backward in loops.
# Either we can move forward or backward in loops but both doesn't work at same time as like recursion so we are using recursion
```

```python
def greet():
    print("Hi")
    greet()
greet() # stack overflows, maximum recursion limit is exceeded
```

```python
# return statement goes back to caller 
def greet(n):
    if(n>5):
        return None
    print("Hi")
    greet(n+1)
greet(1)
```

```python
def m1(n):
    if(n>5):
        return None
    print(n)
    m1(n+1)
m1(1)
```

```python
# After recursion (moving forward) it goes backward where the caller is there it is backtracking and removes function calling in stack
```
