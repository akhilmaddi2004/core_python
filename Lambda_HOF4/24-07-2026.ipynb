```python
# 1. Given a list of tuples (name, marks), sort the list:
#     * first by marks (descending)
#     * then by name (ascending)
lst = [("Akhil",90),("Abhilash",84),("Vamshi",70),("Arjun",87)]
print(sorted(lst, key=lambda x:x[1],reverse=True))
print(sorted(lst, key= lambda x:x[0]))
```

```python
# 3. Given a list of integers, filter numbers divisible by both 2 and 5, add 5 to each using map(), then find the product using reduce().
from functools import reduce
l = [2,34,21,5,6,10,290]
l = list(filter(lambda x: x%2 ==0 and x%5==0,l))
print(l)
l = list(map(lambda x:x+5,l))
print(l)
l = reduce(lambda x,y:x*y, l)
print(l)
```

```python
# 4. Given a list of integers, write a program to:

# * Filter numbers divisible by 2 but not by 4
# * Add 3 to each using map()
# * Sort the result in descending order
# * Find the product of all elements using reduce()

l = [2,4,10,24,26,18,14,16]
l = list(filter(lambda x: x%2==0 and x%4!=0,l))
print(l)
l = list(map(lambda x:x+3,l))
print(l)
l = sorted(l, key= lambda x:x,reverse=True)
print(l)
l = reduce(lambda x,y: x*y,l)
print(l)
```

```python
# 5. Given a list of words:

# * Filter words that start and end with the same letter
# * Convert them to lowercase
# * Sort by last character, then length
# * Join all words into a single string using reduce()
words = ["madam","mam","hello","noon","akhil","nun","amma"]
l = list(filter(lambda x:x[0] == x[-1],words))
print(l)
l = list(map(lambda x:x.lower(),l))
print(l)
l = sorted(l,key=lambda x: x[-1])
print(l)
l = sorted(l, key=lambda x: len(x))
print(l)
l = reduce(lambda x,y: x+y,l)
print(l)
```

```python
# 6. Given a list of transactions where each transaction contains a type (credit or debit) and an amount, write a program to filter only the credit transactions, apply a 5% bonus to each transaction amount using map(), sort the updated transactions in descending order based on the amount, and finally compute the total credited amount using reduce().
# INPUT: 
# transactions = [
#     {"type": "credit", "amount": 1000},
#     {"type": "debit", "amount": 500},
#     {"type": "credit", "amount": 2000}
# ]
transactions = [
    {"type": "credit", "amount": 1000},
    {"type": "debit", "amount": 500},
    {"type": "credit", "amount": 2000}
]
credit = list(filter(lambda x: x['type']=='credit',transactions))
print(credit)
bonus = list(map(lambda x: x['amount'] + (x['amount']*0.05),credit))
print(bonus)
desc = sorted(bonus,key=lambda x : x,reverse=True)
print(desc)
l = reduce(lambda x,y: x+y,desc)
print(l)
```
