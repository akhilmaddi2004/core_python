```python
# Lambda (Anonymous Functions)
# a function which doesn't have any name is anonymous function
# a single line function

# Lambda Expression
# It creates a single line function
# here in lambda expression we "lambda" as keyword to create function in single line
# and in lambda we won't use "return" keyword

# DEf:
# A lambda is a small anonymus function return in single line.
# It's useful for short throw away applications/operations especailly when passing a function as an argument

# SYNTAX:
# lambda parameters : expression

# ex:
a = lambda x : x**2
# Function name(a): stores the lambda function so it can be reused later.
# Lambda keyword (lambda) : defines an anonymous (inline) function.
# Argument(x) : input value passed to the lambda function.
# Expression(x**2) : operation performed on the argument and returned as the result


# We can do only single operation, or less code functions but we can't pass loops in lambda and we can pass conditional stmts
```

```python
def square(a):
    return a*a
result = square(5)
print("square using functions:",result)

# where we use only one time in fucntion we use lambda

Square = lambda a : a*a
print(Square(4))
print(Square(6))
res = Square(5)
print("square using lambda:",res)
```

```python
# adding two numbers using lambda
a = lambda a,b : a+b
print(a(2,5))
```

```python
x= int(input())
if(x>0):
    print("Positive")
else:
    if(x<0):
        print("Negative")
    else:
        print("Zero")
```

```python
# checking +ve, -ve, 0 
check = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(check(5))   
print(check(-3))  
print(check(0))
```

```python
# If else in lambda
# syntax:

# variable_name = lambda parameters : True_value if condn else False

# even or odd using lambda
check = lambda x : "even" if x%2 == 0 else "odd"
print(check(34))
print(check(13))
```

```python
# list sorting using lambda
l = [3,7,4,6,29,9,14,6]
l.sort()
print(l)

# ascending order 
l.sort(key= lambda x : x)
print("ascending order",l)

# descending order
l.sort(key= lambda x:-x)
print("descending order",l)

# list sorted by x/2 quoient 
l.sort(key=lambda x: x/2)
print(l)

l.sort(key=lambda x: x//2)
print(l)

# list sorted by x % 2 remainder
l.sort(key=lambda x: x%2)
print(l)
```

```python
# 1. Write a lambda function that takes a number and returns its cube.
cube = lambda x: x ** 3
print(cube(4))
print(cube(5))
```

```python
# 2. Create a lambda that takes two numbers and returns the larger one using a conditional expression (x if x>y else y)

biggest = lambda x,y: x if x>y else y
print(biggest(12,4))
print(biggest(3,5))
```

```python
# 3. Convert this regular function into lambda: def even(n): return n%2 ==0

check = lambda n : "Even" if n%2==0 else "Odd"
print(check(90))
print(check(7))
```

```python
# 4. Use a lambda with .sort() to sort this list of tuples by the second element: [(1,'banana'),(2,'apple'),(3,'cherry')]

fruits = [(1,'banana'),(2,'apple'),(3,'cherry')]
fruits.sort(key = lambda x: x[1])
print(fruits)
```

```python
# 5. Can a lambda function call another function inside it? Write an example.

# Yes,a lambda function can call another function inside its expression, provided the target function is defined or imported before the lambda is executed. 
# This is commonly used to apply transformations or logic within a single expression.

# Define a regular function.
def square(x):
    return x**2

# Lambda function that calls the 'square' function
call_square = lambda x: square(x)

print(call_square(5)) # 25

# Lambda functions can also call other lambda functions (nested lambdas) or built-in functions like len(), str(), or int() directly within their body.  
# However, if the logic becomes complex, it is often better to use a standard def function for readability.
```

```python
# 6. What are the three main limitations of lambda compared to def? List them.

# 1. Single Expression Only: Lambda functions can only contain a single expression and cannot include statements (such as if blocks, for loops, try-except blocks, or multiple lines of code). 

# 2. Anonymity and Debugging: Lambdas are unnamed (anonymous), meaning they lack a descriptive name which makes debugging more difficult and prevents the use of docstrings or type hints for documentation. 

# 3. Limited Reusability: Because they are designed for short, throwaway operations, lambdas have limited reusability and are generally unsuitable for complex logic that requires maintenance or multiple operations.

# A lambda is anonymous, restricted to one expression, and defined inline.
```

```python
# 1. Write a lambda to calculate simple interest.
# Formula: (P * R * T) / 100

SI = lambda P,T,R : (P*T*R)/100
print("Simple interest:",SI(20000,4,2))
```

```python
# 2. Temperature Converter
# Write a lambda to convert Celsius to Fahrenheit.
# Formula: (C * 9/5) + 32
Fh = lambda C : (C*9/5) + 32
print("Temperature in Farenheit:",Fh(80))
```

```python
# 3. Electricity Bill
# Write a lambda that calculates bill amount:

# * If units ≤ 100 → ₹5/unit
# * Else → ₹8/unit

bill_amt = lambda units : units*5 if units<=100 else units*8 
print("Electricity bill(If units ≤ 100 → ₹5/unit):",bill_amt(90))
print("Electricity bill(If units > 100 → ₹8/unit):",bill_amt(190))
```

```python
# 4. Login Check
# Write a lambda that checks if username equals "admin" and password equals "1234" and returns "Login Success" or "Invalid".

login = lambda  username,password : "Login Success" if username == "admin" and password == "1234" else "Invalid"
print(login(username="admin",password="1234"))
print(login("admin","1234"))
print(login("admin","124"))
print(login(username="akhil",password="1234"))
```
