```python
# Higher order functions

# A Higher order function if any function that takes another function as an argument returns a function.
# Python has four built-in higher order functions:

# 1. Map: Apply a function on every element of an iterable
# 2. Filter: keep only elements where a function returns True
# 3. Reduce: Combine all elements into a single value
# 4. Sorted: Sort any iterable using a custom key function.

# 1. Map: map()
l =  [1,2,3,4]
p = list(map(lambda x: x**2, l))
print(p)

# Syntax: map(function, iterable)

l1 = [1,2,3,4]
l2 = [5,6,7,8]
y = list(map(lambda a,b:a+b,l1,l2))
print(y)

s = [3,7,9,11]
h = list(map(lambda x:x/2,s))
print(h)

```

```python
# 2. Filter: filter()
def isEven(x):
    if x%2==0:
        return True
    else:
        return False
l = [17,29,14,9,6,4]
l1 = list(filter(isEven,l))
print(l1)

l2 = [1,2,3,4]
# l2 = list(map(lambda x: x**2, l2))
# print(l2) # squares
# l2 = list(filter(isEven,l2))
# print(l2) #evens

print(list(filter(lambda x: x%2 ==0 ,list(map(lambda x: x**2,l2)))))
```

```python
# 1. An online store stores product prices in a list. Write a program using map() to apply a 10% tax to each product price and display the upadated prices

lst = [199,99,299,279,599]
tax_added = list(map(lambda x: x+(x*0.1),lst))
print(tax_added)
```

```python
# 2. A list of usernames is stored in lowercase. Use map() to format them so that the first letter is uppercase.
usernames = ["akhil","abhilash","vamshi","karthik"]
title = list(map(lambda x:x.title(),usernames))
print(title)

usernames = ["akhil","vamshi","abhilash","karthik"]
title = list(map(lambda x: x.capitalize(), usernames))
print(title)
```

```python
# 3. An e-commerce website wants to display only products priced above ₹500. Use filter() to extract those prices from a list.
prices = [599,499,500,199,1099,999]
fil = list(filter(lambda x: x>500,prices))
print(fil)
```

```python
# 4. Use map() with a lambda function to multiply each number in a list by 5.
lst = [2,3,48,9,23]
lst = list(map(lambda x: x*5, lst))
print(lst)
```

```python
# Write a program that uses map() to calculate the length of each word in a list of strings.

l = ['apple','banana','cat','dog']
l = list(map(lambda x: len(x),l))
print(l)
l = ['apple','banana','cat','dog']
l1=list(map(len,l))
print(l1)
```
