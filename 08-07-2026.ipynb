## short-hand format for decorators

```python
# Implementing two-ways of decorators
# manual decorator
def decorator1(func):
    def wrapper(*args,**kwargs):
        print("before calling")
        func(*args,**kwargs)
        print("after calling")
    return wrapper

def add(a,b):
    print(a+b)

add = decorator1(add)
add(10,20)
```

```python
# short-hand format
def decorator1(func):
    def wrapper(*args,**kwargs):
        print("Before calling")
        func(*args,**kwargs)
        print("After calling")
    return wrapper

@decorator1
def add(a,b):
    print(a+b)

add(10,20)
```

### Section 3 : The @decorator Syntax

```python
# Python provides a cleaner, more readable shorthand for applying decorators: the @ symbol placed on the line immediately above the function definition.

# Syntax
# @decorator syntax :

# @my_decorator
# def my_function():
#     ....

# This is EXACTLY equivalent to writing:

# def my_function():
#     .....
# my_function = my_decorator(my_function)

# The @ is Just Syntax Sugar
# @my_decorator is shorthand. Python automatically calls my_decorator(function) and reassigns the name. Both ways do the same thing - @ is just cleaner and more readable.

# side-by-side comparison
#   #style              # How it Looks                  # Notes
#   #Manual style       # say_hi = decorator(say_hi)    # Explicit but clutters code
#   # @style            # `@decorator` above def        # Clean, readable, Pythonic
```

```python
# Full Example with @syntax
def shout(func):
    def wrapper(*args,**kwargs):
        result = func(*args,**kwargs)
        return str(result).upper()
    return wrapper
@shout
def greet(name):
    return f"Hello! {name}"
print(greet("Akhil"))
```

```python
def shout(func):
    def wrapper(*args,**kwargs):
        result = func(*args,**kwargs)
        # return result.upper()
        x = result.upper()
        return x
    return wrapper
@shout
def greet(name):
    return f"Hello! {name}"
print(greet("Akhil"))
```

```python
# 1.     Create a function get_message() that returns "hello user". Write a decorator using @ syntax that converts the output to uppercase.
def dec1(func):
    def wrapper(*args,**kwargs):
        res = func(*args,**kwargs)
        return res.upper()
    return wrapper
@dec1
def get_message():
    return f"hello user"
print(get_message())
```

```python
# 2.     Create a function get_number() that returns 10
#     Use a decorator to return double the value.
def double(func):
    def wrapper(*args,**kwargs):
        res = func(*args,**kwargs)
        return res*2
    return wrapper
@double
def get_number(n):
    return n
print(get_number(5))
```

```python
# 3.     Create a function place_order(item)
#     Use a decorator to print:
#     * “Order process started”
#     * “Order process completed”
def dec3(func):
    def wrapper(*args,**kwargs):
        print("Order process started")
        func(*args,**kwargs)
        print("Order process completed")
    return wrapper
@dec3
def place_order(item):
    print("You ordered:",item)
place_order("Laptop")
```
