```python
# 1. Given a list of tuples (name, marks), sort the list:
#     * first by marks (descending)
#     * then by name (ascending)
tup = [
    ("Akhil",89),
    ("Arun",78),
    ("Swamy",98),
    ("Sara",67),
    ("Thiru",55),
]
desc = sorted(tup,key=lambda x:x[1],reverse= True)
print(desc)
ascen = sorted(tup,key=lambda x:x[0])
print(ascen)
```

```python
# 2. Given a list of strings, sort them based on:
#     * length of string
#     * and then alphabetically
strs = ["Akhil","Mani","Naresh","Sri Hari","Abhilash","Zendaya"]
print(sorted(strs, key=lambda x:len(x)))
print(sorted(strs,key=lambda x:x))
```

```python
# Given a list of integers, filter numbers divisible by both 2 and 5, add 5 to each using map(), then find the product using reduce().
from functools import reduce
l = [2,10,3,34,57,90,87,88,85]
fil = list(filter(lambda x: x%2 == 0 and x%5 == 0 ,l))
print(fil)
maps = list(map(lambda x:x+5,fil))
print(maps)
red = reduce(lambda x,y:x*y,maps)
print(red)
```

```python
# 4. Given a list of integers, write a program to:

# * Filter numbers divisible by 2 but not by 4
# * Add 3 to each using map()
# * Sort the result in descending order
# * Find the product of all elements using reduce()
integers = [2,3,4,16,34,78,40,98,11,14]
integers = list(filter(lambda x:x%2==0 and x%4!=0, integers))
integers = list(map(lambda x:x+3, integers))
integers = sorted(integers, key=lambda x:x, reverse=True)
integers = reduce(lambda x,y:x*y,integers)
print(integers)
```

```python
# 5. Given a list of words:

# * Filter words that start and end with the same letter
# * Convert them to lowercase
# * Sort by last character, then length
# * Join all words into a single string using reduce()
l = ['mam','class','classic','neon','nun','noon']
fil = list(filter(lambda x: x[0] == x[-1], l))
print(fil)
lower = list(map(lambda x:x.lower(),fil))
print(lower)
sort = sorted(lower,key=lambda x:x[-1])
print(sort)
sort = sorted(sort,key=lambda x:len(x))
print(sort)
red = reduce(lambda x,y:x+y,sort)
print(red)
```

```python
# 6. Given a list of transactions where each transaction contains a type (credit or debit) and an amount, write a program to filter only the credit transactions, apply a 5% bonus to each transaction amount using map(), sort the updated transactions in descending order based on the amount, and finally compute the total credited amount using reduce().
# INPUT: 
# transactions = [
#     {"type": "credit", "amount": 1000},
#     {"type": "debit", "amount": 500},
#     {"type": "credit", "amount": 2000}
# ]
transactions = [
    {"type": "credit", "amount": 1000},
    {"type": "debit", "amount": 500},
    {"type": "credit", "amount": 2000}
]
credit = list(filter(lambda x: x['type']=='credit',transactions))
print(credit)
bonus = list(map(lambda x: x['amount'] + (x['amount']*0.05),credit))
print(bonus)
desc = sorted(bonus,key=lambda x : x,reverse=True)
print(desc)
l = reduce(lambda x,y: x+y,desc)
print(l)
```
