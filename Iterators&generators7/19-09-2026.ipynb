## Dunder Methods

```python
# Dunder means double underscore
# Dunder methods are special methods whose names starts and ends with '2' underscores.
# ex: __str__(), __eq__(), __add__()
# They allow python to know how objects should behave with built-in functions.

# 1. __str__(): It defines a readable string representation of an object.
class A:
    pass
obj = A()
print("A:",obj) # returns object address

class A:
    def __str__(self):
        return "hi"
obj = A()
print("str:",obj) #hi (overriding str())

# 2. __repr__(): It provides more informative representation of a object.
# 3. __len__(): It tells python what value should be returned when len(obj).
# 4. __contains__(): It defines what should happen when we use membership operator.

# Arithmetic operator dunder methods:
# These methods allow us to defines how operators such as +,-,*,/,//,%,**.
# This is known as operator overloading(customizing operator behavior).
# operator      # Dunder method
#   +           # __add__()
#   -           # __sub__()
#   *           # __mul__()
#   /           # __truediv__()
#   //          # __floordiv__()
#   %           # __mod__()
#   **          # __pow__()

l = [1,2,3,4,5,6,7,8]
print("sum:",sum(x for x in l))
print("even_sum:",sum(x for x in l if(x%2==0)))
```

```python
class A:
    def __init__(self,n):
        self.n = n
obj = A(10)
print("obj address:",obj)

class Student:
    def __init__(self,name,age,marks,subjects):
        self.name = name
        self.age = age
        self.marks = marks
        self.subjects = subjects
    def __str__(self):
        # return f"{self.name} has gotten {self.marks}"
        return f"Hey! This is {self.name}"
    def __repr__(self):
        return f"Student({self.name},{self.age},{self.marks},{self.subjects})"
    def __len__(self):
        return len(self.subjects)
    def __contains__(self, item):
        return item in self.subjects
    def __add__(self, other):
        return self.age + other.age
    def __sub__(self, other):
        return self.marks-other.marks
    def __mul__(self, other):
        return self.age*other.age
    def __truediv__(self, other):
        return len(self.subjects)/len(other.subjects)
    def __floordiv__(self, other):
        return len(self.subjects)//len(other.subjects)
    def __lt__(self, other):
        return self.marks < other.marks
    def __gt__(self, other):
        return self.marks >other.marks
    def _le__(self,other):
        return self.marks >= other.marks
    def __ge__(self, other):
        return self.marks <= other.marks
    def __eq__(self, other):
        return self.marks == other.marks
    def __ne__(self, other):
        return self.marks != other.marks
obj1 = Student("Alice",27,78,['Maths','Science'])
obj2 = Student("Reene",23,79,['Maths','Science','Geography','Civics'])
obj3 = Student("Mag",25,72,['Maths','Science','Social'])
obj4 = Student("Candice",26,98,['Maths'])
obj5 = Student("Max",23,88,['Python','Java'])
print("1.__str__():")
print("Hi",str(obj1))
print(obj1)
print(str(obj1))
print(obj2)
print(obj3)
print("2.__repr__():")
print(repr(obj1))
# we can't add objects without dunder methods
print("3.__len__():")
print("length of subjects of obj1:",len(obj1))
print("length of subjects of obj2:",len(obj2))
print("4.__contains__():")
print("Maths" in obj1)
print("English" in obj1)
print("ArithmeticOperators:")
print("5.__add__():")
print(obj1+obj2)
print("6.__sub__()")
print(obj4-obj3)
print("7.__mul__()")
print(obj1*obj2)
print("8.__truediv__()")
print(obj1/obj2)
print("9.__floordiv__()")
print(obj1//obj2)
print(type(obj1//obj2))
print("Comparison operators:")
print("10.__lt__():")
print(obj1 < obj2)
print("11.__gt__():")
print(obj1 > obj2)
print("12.__le__():")
print(obj1 <= obj2)
print("13.__ge__():")
print(obj1 >= obj2)
print("14.__eq__():")
print(obj1 == obj2)
print("15.__ne__():")
print(obj1 != obj2)
```

```python
class number:
    def __init__(self,n):
        self.n = n
    def __str__(self):
        return f"{self.n}"
    def __add__(self, other):
        return number(self.n + other.n)
    def __sub__(self, other):
        return number(self.n - other.n)
    def __mul__(self, other):
        return number(self.n * other.n)
    def __truediv__(self, other):
        return number(self.n/other.n)
    def __floordiv__(self, other):
        return number(self.n//other.n)
    def __pow__(self, other):
        return number(self.n**other.n)
    def __pow1__(self,power,modulo = None):
        return number(self.n**power)
n1 = number(10)
n2 = number(20)
print("__str__()")
print(n1)
print(n2)
print("__add__()")
print(n1+n2)
print(type(n1+n2))
print("__sub__()")
print(n1-n2)
print("__mul__()")
print(n1*n2)
print("__truediv__()")
print(n1/n2)
print("__floordiv__()")
print(n1//n2)
print("__pow__()")
print(n1**n2)
print(hash(obj1))
```
