## Iterators and Generators

## Generators

```python
# Yield keyword - works same as return but we can write as many time as
# A generator is a special kind of iterator created with a regular function that uses yield keyword instead of return.
# Python handles all ther iter and next machinery for you automatically.
# Generators are also called lazy iterators
def m1():
    yield 10
    yield 20
    yield 30
    yield 40
x = m1()
# for i in x:
#     print(i)
print(next(x))
print(next(x))
print(next(x))
print(next(x))
# print(next(x)) # StopIteration
y = m1()
print()
for i in y:
    print(i)
```

```python
def m1(start,stop):
    current = start
    while(current<=stop):
        yield current
        current += 1
x = m1(1,5)
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print()
y = m1(1,5)
for i in y:
    print(i)
```

```python
def m1(start,stop):
    while(start<=stop):
        yield start
        start += 1
x = m1(5,10)
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
# print(next(x))
print()
y = m1(20,30)
for i in y:
    print(i)
```

```python
# we can control infinite loop with generators
def m1():
    n = 1
    while True:
        yield n
        n += 1
x = m1()
print(next(x))
print(next(x))
for _ in range(1,5):
    print(next(x))
```

```python
def g2():
    n = 1
    while True:
        yield n
        n += 1
x = g2()
print(next(x))
print(next(x))
print(next(x))
print(next(x))
y = g2()
print()
for _ in range(1,6):
    print(next(y))
```

```python
def fibonacci():
    a,b = 0,1
    while True:
        yield a
        a,b = b,a+b 
x = fibonacci()
print(next(x))
print(next(x))
print(next(x))
print()
y = fibonacci()
for _ in range(1,6):
    print(next(y))
```

```python
# comprehensions
l = list(range(1,10))
print(l)
# list comprehension
l = [x for x in range(1,10)]
print(l)
print(type(l))
l = [x*x for x in range(1,10)]
print(l)
# tuple comprehension
t = (x for x in range(1,10)) # it returns generator expression
print(t)
print(type(t))
# so do type conversion
t = tuple(x for x in range(1,10))
print(t)
print(type(t))
# set
s = {x for x in range(1,10)}
print(s)
print(type(s))
# dict
d = {x:x for x in range(1,10)}
print(d)
print(type(d))
d = {x:x*x for x in range(1,10)}
print(d)
print(type(d))

t = (x for x in range(1,5))
for i in t:
    print(i,end=" ")
print()
g = (x if x%2==0 else x+1 for x in range(1,5))
for i in g:
    print(i,end=" ")
print()
g = (x if x%2==0 else "" for x in range(1,5))
for i in g:
    print(i,end=" ")
print()
g2 = (x for x in range(1,10) if(x%2==0))
for i in g2:
    print(i,end=" ")
print()
print(max(x for x in range(1,10)))
print(max(x for x in range(1,10) if(x%2==0)))
print(min(x for x in range(1,10) if(x%2==0)))
print(min(x for x in range(1,10)))
print(sum(x for x in range(1,10) if(x%2==0)))
print(sum(x for x in range(1,10)))
import sys
print()
l = [x for x in range(100000)]
print(sys.getsizeof(l))
h = (x for x in range(100000))
print(sys.getsizeof(h))
```
