### Iterators problems

```python
# 1.	Create an custom iterator that prints numbers from 1 to N, where N is given by the user.
class Numbers:
    def __init__(self,n):
        self.n = n
        self.start = 1
    def __iter__(self):
        return self
    def __next__(self):
        if(self.start<=self.n):
            num = self.start
            self.start += 1
            return num
        else:
            raise StopIteration
n1 = Numbers(int(input()))
for i in n1:
    print(i,end=" ")
```

```python
# 2.	Create an custom iterator that prints numbers from N to 1.
class Down:
    def __init__(self,n):
        self.n = n
    def __iter__(self):
        return self
    def __next__(self):
        if(self.n>=1):
            num = self.n
            self.n -= 1
            return num
        else:
            raise StopIteration
d1 = Down(int(input()))
for i in d1:
    print(i,end=" ")
```

```python
# 3.	Create an custom iterator that prints the first N even numbers.
class Nevens:
    def __init__(self,n):
        self.n = n
        self.count = 0
        self.start = 0
    def __iter__(self):
        return self
    def __next__(self):
        if(self.count<self.n):
            num = self.start
            self.start += 1
            if(num%2==0):
                self.count += 1
                return num
            else:
                return next(self)
        else:
            raise StopIteration
e1 = Nevens(int(input()))
for i in e1:
    print(i,end=" ")
```

```python
def digits(n):
    result = []
    while(n>0):
        r = n%10
        result.append(r)
        n = n//10
    index = len(result)-1
    while(index>=0):
        yield result[index]
        index -=1
d = digits(123456)
for i in d:
    print(i,end=" ")
```

```python
# 9.	Create a generator that yields cumulative sum of numbers in a list. Example: [1,2,3] → 1, 3, 6
def cul_sum(n):
    total = 0
    for i in n:
        total += i
        yield total
l = cul_sum([1,2,3])
for i in l:
    print(i,end=" ")
```

```python
def cul_sum(n):
    total = 0
    index = 0
    while index<len(n):
        total += n[index]
        yield total
        index += 1
l = cul_sum([10,20,30])
for i in l:
    print(i,end=" ")
```

```python
# 1. Create a class called Product with attributes name and price. Change the + operator so that adding two Product objects returns the sum of their prices.

# p1 = Product("Keyboard", 1500)
# p2 = Product("Mouse", 700)
# print(p1 + p2)

# Expected Output:
# 2200

# Operator to change: +
# Magic Method: _add_()
class Product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    def __add__(self, other):
        return self.price + other.price
p1 = Product("Keyboard",1500)
p2 = Product("Mouse",700)
print("adding two Product Prices:",p1+p2)
```

```python
# 2. Create a class called BankAccount with an attribute balance. Change the - operator so that subtracting one account from another returns the difference between their balances.

# account1 = BankAccount(10000)
# account2 = BankAccount(3500)
# print(account1 - account2)

# Expected Output:
# 6500

# Operator to change: -
# Magic Method: _sub_()
class BankAccount:
    def __init__(self,balance):
        self.balance = balance
    def __sub__(self, other):
        return self.balance - other.balance
acc1 = BankAccount(10000)
acc2 = BankAccount(3500)
print("difference between their balances:",acc1-acc2)
```

```python
# __radd__() definition

# __radd__() is a magic method used for reflected addition. It is called when an object is on the right side of + and the left-side object's __add__() cannot handle the operation.
# 7. Create a class called Score with an attribute points. Change the addition behaviour so that an integer can be added on the left side of a Score object.

# score = Score(80)
# print(20 + score)

# Expected Output:
# 100

# Operator to change: Reflected +
# Magic Method: _radd_()
class Score:
    def __init__(self,points):
        self.points = points
    def __radd__(self, other):
        return other + self.points
score = Score(80)
print(20+score)

class Score:
    def __init__(self,points):
        self.points = points
    def __radd__(self, other):
        return self.points + other
score = Score(80)
print(20+score)
```

```python
# __iadd__() definition

# __iadd__() is a magic method used to change the behavior of the += operator. It performs addition and updates the existing object itself.
# 8. Create a class called Inventory with an attribute quantity. Change the += operator so that the supplied quantity is added directly to the existing object.

# stock = Inventory(50)
# stock += 20
# print(stock.quantity)

# Expected Output:
# 70

# Operator to change: +=
# Magic Method: _iadd_()
class Inventory:
    def __init__(self,quantity):
        self.quantity = quantity
    def __iadd__(self, other):
        self.quantity += other
        return self
stock = Inventory(50)
stock += 20
print(stock.quantity)
```

```python
# 9. Create a class called Permission with an attribute containing a set of permission names. Change the | operator so that combining two Permission objects returns all unique permissions from both objects.

# p1 = Permission({"read", "write"})
# p2 = Permission({"write", "delete"})
# result = p1 | p2
# print(result.permissions)

# Expected Output:
# {'read', 'write', 'delete'}

# Operator to change: |
# Magic Method: _or_()
class Permission:
    def __init__(self,permissions):
        self.permissions = permissions
    def __or__(self, other):
        return self.permissions | other.permissions
p1 = Permission({"read","write"})
p2 = Permission({"write","delete"})
result = p1 | p2
print(result)
print(type(result))
class Permission:
    def __init__(self,permissions):
        self.permissions = permissions
    def __or__(self, other):
        return Permission(self.permissions | other.permissions)
p1 = Permission({"read","write"})
p2 = Permission({"write","delete"})
result = p1 | p2
print(result.permissions)
print(type(result))
```

```python
# 10. Create a class called Team with attributes name and points. Change the <= operator so that one team is considered less than or equal to another when its points are lower than or equal to the other team's points.

# team1 = Team("Falcons", 25)
# team2 = Team("Tigers", 30)
# print(team1 <= team2)

# Expected Output:
# True

# Operator to change: <=
# Magic Method: _le_()
class Team:
    def __init__(self,name,points):
        self.name = name
        self.points = points
    def __le__(self, other):
        return self.points <= other.points
t1 = Team("Falcons",25)
t2 = Team("Tigers",30)
print(t1 <= t2)
```
