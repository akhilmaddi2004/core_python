```python
#       Iterable                    # Iterator
# Something you loop over it.       # Something that produces one value at a time.
# Has an iter method that returns   # It has both iter and next
# an iterator
# Ex: list,tuple,dict,set,string,range

# iter() and next():
# - An object that implements both iter() and next() methods correctly is an iterable
# - iter():
# - def __iter__(self): # method signature
# - iter must return an iterator object.
# - It is called by iter(obj) or obj.__iter__() and for loop.
# - next():
# - def __next__(self): # method signature
# - It returns next value and raises StopIteration when values are exhausted.
# - It can be called using next(obj) or obj.__next__() and for loop
```

```python
l = list(range(1,11))
print(l)
print()

# custom iterator(range)
class Numbers:
    def __init__(self,start,end):
        self.val = start
        self.end = end 
    def __iter__(self):
        return self
    def __next__(self):
        if(self.val<=self.end):
            curr = self.val
            self.val += 1
            return curr
        else:
            raise StopIteration
obj = Numbers(1,11)
for i in obj:
    print(i)
print()
fivetoten = Numbers(5,10)
for i in fivetoten:
    print(i)
print()

twenty_to_hundred = Numbers(20,100)
for i in twenty_to_hundred:
    print(i)
```

```python
# custom iterator(fibonacci)
class fibonacci:
    def __init__(self,n):
        self.count = 0
        self.n = n
        self.a,self.b = 0,1
    def __iter__(self):
        return self
    def __next__(self):
        if(self.count<=self.n):
            val = self.a
            self.a,self.b = self.b, self.a+self.b
            self.count += 1
            return val
        else:
            raise StopIteration
five = fibonacci(5)
for i in five:
    print(i)
print()
ten = fibonacci(10)
for i in ten:
    print(i)
```

```python
# prime
def is_prime(n):
    if(n<2):
        return False
    for i in range(2,int(n**0.5)+1):
        if(n%i==0):
            return False
    return True
class prime:
    def __init__(self,n):
        self.n = n
        self.curr = 2
    def __iter__(self):
        return self
    def __next__(self):
        while(self.curr<=self.n):
            num = self.curr
            self.curr += 1
            if(is_prime(num)):
                return num
        raise StopIteration
x = prime(20)
for i in x:
    print(i)
print()
y = prime(5)
print(next(y))
print(next(y))
print(next(y))
# print(next(y))
```
