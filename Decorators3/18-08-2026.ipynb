# LEGB Rule in python

```python
# LEGB stands for Local, Enclosing, Global, and Built-in.
# It defines the order in which Python searches for a variable or name.
```

```python
# 1. L -> Local 
# A local variable is a variable created inside the current function.
# Python checks the local scope(local memory) first because it belongs directly to the function being executed.

# Reason : To find variables that are directly avaliable inside the current function.

def test():
    x = 10
    print(x)
test()

# Here, x is Local to test()
```

```python
# 2. E -> Enclosing
# The enclosing scope is the scope of an outer function when another function is defined inside it.
# Python checks this after Local when the variable is not found in the current function.
# This happens with nested functions.

# Reason: To allow an inner function to access variables from its outer function.

def outer():
    x = 20
    def inner():
        print(x)
    inner()
outer()
# Here, x is Enclosing for inner() because it belongs to outer()
```

```python
# 3. G -> Global
# A global variable is defined outside all functions and classes and can generally be accessed throughout the program.
# Python checks the Global scope after Local and Enclosing scopes.

# Reason: To access variables that are avaliable at the program/module level.

x = 30
def test():
    print(x)
test()

# Here, x is Global because it is defined outside the function.
```

```python
# 4. B -> Built-in
# Built-in names are provided by Python itself, such as print(), len(), sum(), and max().
# Python checks the Built-in scope last if the name is not found in the other scopes.

# Reason: To provide access to Python's predefined functions and objects.

def test():
    print(len("Hello"))
test()
# Here, len() is found in the Built-in scope.
```

```python
# | Letter | Scope     | Where?            | Reason                                  |
# | ------ | --------- | ----------------- | --------------------------------------- |
# | **L**  | Local     | Current function  | Directly available variables            |
# | **E**  | Enclosing | Outer function    | Inner function accesses outer variables |
# | **G**  | Global    | Outside functions | Program-level variables                 |
# | **B**  | Built-in  | Python itself     | Predefined Python functions/objects     |


# Search order:
# L -> E -> G -> B
# Local → Enclosing → Global → Built-in
# Imp: Python searches in this order and stops as soon as it finds the name.
```

```python
# 1. Local
def m1():
    x = 10 # local variable/ local memory space
    print(x)
m1()
```

```python
x = 5 
def m1():
    x = 10
    print(x)
m1()
print(x)
# The variable inside function we can only used in inside the function.
# we can only access local variable in m1 func memory scape only and we can't use it in out of the function
```

```python
# 2. Enclosing:
# checking variable in outer function
# it checks firstly in inner func then goes to outer function , enclosing is useful in nested function/closures
def m2():
    x = 5
    def m1():
        # x = 10
        print(x)
    m1()
m2()
```

```python
def m2():
    x = 5
    def m1():
        x = 10
        print(x)
    m1()
m2()
```

```python
# Global - here in global it is declared out of the all functions
def say_hi():
    print("Hi")

x = 14
func = say_hi

def m2():
    def m1():
        # x = 10
        print(x)
        func()
    m1()
m2()
# here if we modify 'x' in our inner function the original global 'x' doesn't updates 
```

```python
# By using global x keyword the global variable can be modified by accessing in local memory space.
# If we won't use global keyword then in it only changed in local memory space but not in global variable
x = 14
def m1():
    global x
    x = 5
    print(x)
m1()
print(x)
```
