### Triple Layer Decorator

```python
# Triple level nesting
import functools
def dec1(args1):
    def dec(func):
        @functools.wraps(func)
        def wrap1(*args,**kwargs):
            for i in range(args1):
                func(*args,*kwargs)
        return wrap1
    return dec
@dec1(3)
def say_hello():
    print("Hello Akhil")
say_hello()
```

```python
# Understand this important line
# say_hello = dec1(3)(say_hello)

# Break it into two steps:

# Step 1
# dec1(3)

# returns:

# dec

# So:

# dec1(3)(say_hello)

# means:

# dec(say_hello)

# And dec(say_hello) returns:

# wrap1

# Therefore:

# say_hello = wrap1

# Then:

# say_hello()

# actually calls wrap1().

# Inside wrap1():

# for i in range(3):
#     func()

# So the original say_hello() runs 3 times.
```

```python
import functools
def dec1(args1):
    def dec(func):
        @functools.wraps(func)
        def wrap1(*args,**kwargs):
            for i in range(args1):
                result = func(*args,**kwargs)
            return result
        return wrap1
    return dec
def say_hello():
    print("hello")
say_hello = dec1(3)(say_hello)
say_hello()
```

```python
# 3. Create a function process_order(price, discount_function).

# Pass different lambda functions to calculate:

# * 10% discount
# * 20% discount
# * ₹100 flat discount

# Create two decorators:

# * order_logger → logs the order processing
# * payment_check → prints "Payment verification completed"

# Apply both decorators to the function.

def order_logger(func):
    def wrap1(*args,**kwargs):
        print("logs the order processing")
        result = func(*args,**kwargs)
        return result
    return wrap1

def payment_check(func):
    def wrap2(*args,**kwargs):
        result = func(*args,**kwargs)
        print("Payment verification completed")
        return result
    return wrap2

discount_10 = lambda x: x-(x*0.1)
discount_20 = lambda x: x-(x*0.2)
discount100_off = lambda x: x-100

@payment_check
@order_logger
def process_order(price,discount_function):
    return discount_function(price)

print(process_order(1099,discount_10))
print(process_order(1099,discount_20))
print(process_order(1099,discount100_off))

@payment_check
@order_logger
def process_order(price,discount_function):
    return list(map(discount_function,price))
price = [1000,2000,300,399]
print()
print(process_order(price,discount_10))
print(process_order(price,discount_20))
print(process_order(price,discount100_off))
```

```python
# 4 . Create a function send_notification(message, formatter).

# Use lambda functions as formatter to:

# * Convert the message to uppercase
# * Convert the message to lowercase
# * Add "!!!" to the message

# Create a decorator that prints "Notification started" before execution and "Notification sent" after execution.
def dec(func):
    def wrap1(*args,**kwargs):
        print("Notification started")
        result = func(*args,**kwargs)
        print("Notification sent")
        return result
    return wrap1

upper = lambda x: x.upper()
lower = lambda x: x.lower()
# add = lambda x: x+str("!!!")
add = lambda x: x+"!!!"

@dec
def send_notification(message,formatter):
    return formatter(message)

print(send_notification("Hii, Akhil",upper))
print(send_notification("Hii, Akhil",lower))
print(send_notification("Hii, Akhil",add))
```
