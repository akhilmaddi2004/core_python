```python
# Function Decorators - Built from Scratch

# Before using the fancy @syntax, let's build a decorator manually step by step. This way you will truly understand what is happening under the hood.
```

```python
# Step 1 - Functions Are Objects
# In Python, Functions are first-class objects. You can store them in variable

# Functions stored in variables:
def greet():
    print("Hello!")
say_hello = greet # say_hello now points to the same function
say_hello() # Output: Hello!
```

```python
# Step 2 - A function Inside a Function (Closure) (inner function)
# A function can be defined inside another function. The function can access variables from the outer function's scope - this is called a closure.

# inner function example:
def outer():
    message = 'I am the outer function'
    def inner():
        print(message) # inner can see outer's variables
    inner() # call inner from inside outer
outer() 
```

```python
# Step 3 - Returning a Function
# A function can return another function as its result. This is the heart of how decorators work.

# Return a function:
def make_greeter():
    def say_hi():
        print("Hi there!")
    return say_hi # return the function itself, NOT say_hi()

my_func = make_greeter() # my_func is now say_hi
my_func() # calling it
```
