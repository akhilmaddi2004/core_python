##### multiple decorators execution 

```python
#     Create a function apply_discount(price)
#     Use a decorator to print:
#     * “Applying discount…”
#     * “Discount applied”
def decorator1(func):
    def wrapper(*args,**kwargs):
        print("Applying discount...")
        func(*args,**kwargs)
        print("Discount applied")
    return wrapper

@decorator1
def apply_discount(price):
    discount = 50 
    price = price - discount
    print(price)

apply_discount(2500)
```

```python
# @dec2 
# @dec1
# here the dec1 decroator will execute first then dec2 will execute because the execution order of decorators occurs in bottom-top when we have more than 1 decorator
```

```python
def dec2(func):
    def wrapper2(*args,**kwargs):
        print("Hi! User")
        func(*args,**kwargs)
        print("Discount applied")
    return wrapper2
def dec1(func):
    def wrapper1(*args,**kwargs):
        print("Applying discount..")
        func(*args,**kwargs)
        print("discount applied")
    return wrapper1
@dec2
@dec1
def apply_discount(price):
    discount = 50
    price = price - discount
    print(price)
apply_discount(1000)

# dec1(decorator1) takes apply_discount as argument
# Then dec2(decorator2) takes the function with the combination of apply_discount(org_function) & dec1 functions
```

```python
# so when we call as => print(apply_discount.__name__) we get wrapper or wrapper1 or 2
# to get original func_name in decorators we have to import 
# functools module = import functools
# then in after declaration of decorator function use 
# @functools.wraps(func)
# now if we call as => print(apply_discount.__name__) we get apply_discount
# to restore original function we these code lines
```

```python
import functools
def decorator2(func):
    @functools.wraps(func)
    def wrapper2(*args,**kwargs):
        print("Hi! User")
        func(*args,**kwargs)
        print("Discount applied")
    return wrapper2 
def decorator1(func):
    @functools.wraps(func)
    def wrapper(*args,**kwargs):
        print("Applying discount...")
        func(*args,**kwargs)
        print("discount applied")
    return wrapper
@decorator2
@decorator1
def apply_discount(price):
    discount = 30
    price = price - discount
    print(price)
apply_discount(300)
```

```python
print(apply_discount.__name__)
# we got original function name because we have used functools
```
