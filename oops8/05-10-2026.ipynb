## Encapsulation


```python
# Encapsulation in Python is an object-oriented principle that bundles data (attributes) and methods into a single unit(a class) to protect internal state and control access.
# It prevents accidental data modification by restricting direct access to variables, ensuring that changes occur only through defined methods.

# Python implements encapsulation through naming conventions rather than strict access modifiers, using name mangling to enforce privacy:

# -> Public Members: Accessible from anywhere with no prefix(e.g: self.name)
# -> Protected Members: Indicated by a single underscore prefix (e.g: self._salary);
# intended for internal use and subclasses, but accessible externally.
# -> Private Members: Indicated by a double underscore prefix (e.g: self.__balance);
# strictly restricted to the class itself, making them difficult to access from outside.

# Encapsulation is typically realized using getter and setter methods or Python's decorator to validate and control how data is read or modified.
# For example, a private attribute __salary can be exprosed via a get_salary() method, allowing the class to enforce validation rules before updating the value.
```

```python
# Encapsulation
# Encapsulation is a process of granting access to only authorized users for private data or hidden data.
# Encapsulation is a process of protecting the data within a class and restricting direct access to the data.
# Python mainly uses underscores to indicate access levels.
# 1. public access -> self.name / name(cls_var) -> no restriction, can be accessed/modified directly
# 2. protected -> self._age/ _age(cls_var) -> only that class and subclasses
# 3. private -> self.__phone_number(undergoes name mangling) / __phone_number -> direct access only in that class 

# to access private data out of the class we use syntax:
# objName._className__variableName

# @property:
# @property allows us to access a private variable using normal attribute syntax
class Product:
    def __init__(self,price):
        self.__price = price
    @property
    def price(self):
        return self.__price
product1 = Product(100)
print(product1.price) # being able to access a private variable like a normal variable
print()
# @property.setter:
# @property.setter is used to modify a private variable with modification
class Product:
    __profit = 1000
    def __init__(self,price):
        self.__price = price
    @property
    def price(self):
        return self.__price
    @price.setter
    def price(self,new_price):
        if(new_price>0):
            self.__price = new_price
product1 = Product(100)
print(product1.price)
product1.price = 150
print(product1.price)
print(product1._Product__profit)
```

```python
# Encapsulation - Practice Questions

# 9. @property + @setter
# Design a BankAccount class with a private __balance.
# Create @property balance as a getter. 
# Create a @balance.setter that raises ValueError if the value is negative.
# Add methods deposit(amount) and withdraw(amount) that update balance through the setter. 
# Test all edge cases.

class BankAccount:
    def __init__(self,balance):
        self.__balance = balance
    @property
    def balance(self):
        return self.__balance
    @balance.setter
    def balance(self,new_balance):
        if(new_balance>=0):
            self.__balance = new_balance
        else:
            raise ValueError("Balance can't be Negative")
    def deposit(self,amount):
        if(amount>0):
            self.__balance += amount
            # return self.__balance
        else:
            return ValueError("Deposit amount must be positive")
    def withdraw(self,amount):
        if(amount>0 and amount <= self.__balance):
            self.__balance -= amount
            # return self.__balance
        else:
            return ValueError("Invalid withdrawal amount")
b1 = BankAccount(10000)
print("curr_bal:",b1.balance)
# b1.balance = -9000
b1.balance = 9000
print("curr_bal (new-balance):",b1.balance)
b1.deposit(1001)
print("curr_bal (deposit):",b1.balance)
b1.withdraw(1000)
print("curr_bal (withdraw):",b1.balance)
```

```python
# 10. Computed Properties
# Create a class Circle with a private __radius. 
# Add @property radius (getter) and @radius.setter that validates the radius is positive. 
# Add a computed @property area and @property circumference. 
# Show that changing radius automatically updates both computed properties.
import math
class Circle:
    def __init__(self,radius):
        self.__radius = radius
    @property
    def radius(self):
        return self.__radius
    @radius.setter
    def radius(self,new_radius):
        if(new_radius>0):
            self.__radius = new_radius
        else:
            raise ValueError("Radius must be positive")
    @property
    def area(self):
        return math.pi * self.__radius**2
    @property
    def circumference(self):
        return 2 * math.pi * self.__radius
c1 = Circle(5)
print("Before modification:")
print("radius:",c1.radius)
print("area:",c1.area)
print("circumference:",c1.circumference)
print()
print("After modifying radius:")
c1.radius = 6
print("radius (new_radius):",c1.radius)
print("area:",c1.area)
print("circumference:",c1.circumference)
```

```python
class Person:
    def __init__(self,first_name,last_name):
        self.__first_name = first_name
        self.__last_name = last_name
    @property
    def first_name(self):
        return self.__first_name
    @first_name.setter
    def first_name(self,value):
        self.__first_name = value
    @property
    def last_name(self):
        return self.__last_name
    @last_name.setter
    def last_name(self,value):
        self.__last_name = value
    @property
    def full_name(self):
        return self.__first_name + " " + self.__last_name
    @full_name.setter
    def full_name(self,value):
        first,last = value.split()
        self.__first_name = first
        self.__last_name = last
p = Person("Maddi","Akhil")
print(p.first_name)  
print(p.last_name)  
print(p.full_name)  
print()
p.full_name = "Akhil Maddi"
print(p.first_name)
print(p.last_name)
print(p.full_name)
```
