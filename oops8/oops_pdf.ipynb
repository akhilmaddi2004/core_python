```python
# 1. Inheritance + Abstraction.
# INHERITANCE + ABSTRACTION: 
# Design an abstract class Vehicle(ABC) with abstract methods start(), stop(), and fuel_type (abstract property). 
# Create ElectricVehicle and PetrolVehicle as concrete classes. 
# Then create HybridCar(ElectricVehicle, PetrolVehicle) using multiple inheritance. 
# Use super() correctly at each level and print the MRO.

from abc import ABC,abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self): pass
    @abstractmethod
    def stop(self): pass
    @property
    @abstractmethod
    def fuel_type(self): pass

class ElectricVehicle(Vehicle):
    def start(self):
        print("Electric vehicle started")
    def stop(self):
        print("Electric vehicle stopped")
    @property
    def fuel_type(self):
        return "Electricity"

class PetrolVehicle(Vehicle):
    def start(self):
        print("Petrol vehicle started")
    def stop(self):
        print("Petrol vehicle stopped")
    @property
    def fuel_type(self):
        return "Petrol"

class HybridCar(ElectricVehicle,PetrolVehicle):
    def start(self):
        print("Hybrid car starting")
        super().start()
    def stop(self):
        print("Hybrid car stopping")
        super().stop()
    @property
    def fuel_type(self):
        return "Electricity + Petrol"

# create objects
e = ElectricVehicle()
p = PetrolVehicle()
h = HybridCar()

print("Electric Vehicle:")
e.start()
e.stop()
print("Fuel:",e.fuel_type)

print("\nPetrol Vehicle:")
p.start()
p.stop()
print("Fuel:",p.fuel_type)

print("\nHybrid Car:")
h.start()
h.stop()
print("Fuel:",h.fuel_type)

print("\nMRO:")
print(HybridCar.mro())
```

```python
# Polymorphism + Encapsulation 
# POLYMORPHISM + ENCAPSULATION: 
# Build an Employee hierarchy: Employee → Manager, Developer, Intern. 
# Each has private __salary and a @property salary. 
# Each overrides a method monthly_cost() differently (manager gets bonus, intern gets a cap). 
# Write a function total_payroll(employees) that works polymorphically.

class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.__salary = salary
    @property
    def salary(self):
        return self.__salary
    def monthly_cost(self):
        return self.salary

class Manager(Employee):
    def monthly_cost(self):
        bonus = self.salary * 0.20
        return self.salary + bonus

class Developer(Employee):
    def monthly_cost(self):
        allowance = 3000
        return self.salary + allowance

class Intern(Employee):
    def monthly_cost(self):
        if(self.salary>10000):
            return 10000
        return self.salary

# polymorphism function
def total_payroll(empolyees):
    total = 0
    for emp in empolyees:
        cost = emp.monthly_cost()
        print(emp.name,":",cost)
        total += cost
    return total

# creating employee objects
e1 = Manager("Ravi",50000)
e2 = Developer("Akhil",30000)
e3 = Intern("Rahul",12000)

# store objects in a list
employees = [e1,e2,e3]

# Calculate total payroll
print("Total Payroll:",total_payroll(employees))
```

```python
# All Four Pillars 
# ALL FOUR PILLARS: 
# Design a Media Streaming system. 
# Create an abstract class MediaPlayer(ABC) with abstract methods play(), pause(), stop(). 
# Encapsulate __volume with @property and @setter (0–100 validation). 
# Use inheritance: AudioPlayer and VideoPlayer extend MediaPlayer. 
# Demonstrate polymorphism by putting both in a list and calling play() on each.

from abc import ABC,abstractmethod

class MediaPlayer(ABC):
    def __init__(self):
        self.__volume = 50
    @property
    def volume(self):
        return self.__volume
    @volume.setter
    def volume(self,value):
        if(0<=value<=100):
            self.__volume = value
        else:
            print("Invalid volume! Enter between 0 and 100")
    @abstractmethod
    def play(self): pass
    @abstractmethod
    def pause(self): pass
    @abstractmethod
    def stop(self): pass

class AudioPlayer(MediaPlayer):
    def play(self):
        print("Playing audio")
    def pause(self):
        print("Audio paused")
    def stop(self):
        print("Audio stopped")

class VideoPlayer(MediaPlayer):
    def play(self):
        print("Playing video")
    def pause(self):
        print("Video paused")
    def stop(self):
        print("Video stopped")

# create objects
a = AudioPlayer()
v = VideoPlayer()

# change volume
a.volume = 80
v.volume = 60

# Display volume
print("Audio Volume:",a.volume)
print("Video Volume:",v.volume)

# Polymorphism
players = [a,v]
for player in players:
    player.play()
    player.pause()
    player.stop()
```
