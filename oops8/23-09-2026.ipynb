```python
# multiple inheritance
# constructor chaining
# (MRO) # C(child class) -> A(parent class) -> B(intermediate class) -> object
class A:
    pass
class B:
    pass
class C(A,B):
    pass

class A:
    def __init__(self):
        print("A Constructor")
        super().__init__()
class B:
    def __init__(self):
        print("B Constructor")
class C(A,B):
    def __init__(self):
        print("C Constructor")
        super().__init__()
c = C()
print(C.mro())
print(C.__mro__)
```

```python
# constructor chaining
# multi-level inheritance
class A:
    def __init__(self):
        print("A Constructor!!")
class B(A):
    def __init__(self):
        print("B Constructor!!")
        super().__init__()
class C(B):
    def __init__(self):
        print("C Constructor!!")
        super().__init__()
c = C()
print(A.mro())
print(B.mro())
print(C.mro())
b = B()
```

```python
# Hybrid inheritance
class A:
    def __init__(self):
        print("A Constructor!!")
class B(A):
    def __init__(self):
        print("B Constructor!!")
        super().__init__()
class C(A):
    def __init__(self):
        print("C Constructor!!")
        super().__init__()
class D(B,C):
    def __init__(self):
        print("D Constructor!!")
        super().__init__()
d = D()
print(D.mro())
y = (x.__name__ for x in D.mro())
for i in y:
    print(i)
```

```python
# method chaining
class A:
    def __init__(self):
        print("A Constructor!!")
    def m1(self):
        print("M1-A")
    @classmethod
    def m2(cls):
        print("class method in A")
class B(A):
    def __init__(self):
        print("B Constructor!!")
        super().__init__()
    def m1(self):
        print("M1-B")
        super().m1()
    @classmethod
    def m2(cls):
        print("class method in B")
        super().m2()
class C(B):
    def __init__(self):
        print("C Constructor!!")
        super().__init__()
    def m1(self):
        print("M1-C")
        super().m1()
    @classmethod
    def m2(cls):
        print("class method in C")
        super().m2()
c = C()
print()
c.m1()
print()
c.m2()
```

```python
# create a class employee with instance variables name and salary
# create a child class manager that has another variable called team size
# initialize variables while creating object of manager class
class employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
class manager(employee):
    def __init__(self,name,salary,team_size):
        self.team_size = team_size
        super().__init__(name,salary)
    def display(self):
        print("Emp_name:",self.name)
        print("Salary:",self.salary)
        print("team-size:",self.team_size)
m1 = manager("Akhil",200000,5)
m1.display()
print(m1.name)
print(m1.salary)
print(m1.team_size)
```

```python
# A company maintains employee information using seperate classes 
# 1. An employee class stores emp_id and emp_name
# while 2. attendance class stores days present 
# create a manager class that inherits both employee and attendance and additionally adds team size variable
# initialize all values when creating manager object
# display all inforamtion
class employee:
    def __init__(self,emp_id,emp_name,present_days):
        self.emp_id = emp_id
        self.emp_name = emp_name
        super().__init__(present_days)
class Attendance:
    def __init__(self,present_days):
        self.present_days = present_days
class Manager(employee,Attendance):
    def __init__(self,emp_id,emp_name,present_days,team_size):
        self.team_size = team_size
        super().__init__(emp_id,emp_name,present_days)
    def display(self):
        print("Emp_id:",self.emp_id)
        print("Emp_name:",self.emp_name)
        print("Present Days:",self.present_days)
        print("Team-size:",self.team_size)
m1 = Manager(1,"Akhil",10,4)
m1.display()
print(m1.emp_id)
print(m1.emp_name)
print(m1.present_days)
print(m1.team_size)
```

```python
class Employee:
    def __init__(self,emp_id,emp_name):
        self.emp_id = emp_id
        self.emp_name = emp_name
class Attendance:
    def __init__(self,present_days):
        self.present_days = present_days
class Manager(Employee,Attendance):
    def __init__(self,emp_id,emp_name,present_days,team_size):
        self.team_size = team_size
        Employee.__init__(self,emp_id,emp_name)
        Attendance.__init__(self,present_days)
m1 = Manager(101,"Akhil",10,5)
print(m1.emp_id)
print(m1.emp_name)
print(m1.present_days)
print(m1.team_size)
```
