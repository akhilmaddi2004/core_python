```python
# 13

# ABC + @abstractmethod
# Design an abstract class PaymentGateway(ABC) with abstract methods:
# initiate(amount), verify(transaction_id), and refund(transaction_id). Implement two
# concrete classes RazorpayGateway and PayPalGateway. Verify that you cannot
# instantiate PaymentGateway directly.
from abc import ABC,abstractmethod
class PaymentGateway(ABC):
    @abstractmethod
    def initiate(self,amount):
        pass
    @abstractmethod
    def verify(self,transaction_id):
        pass
    @abstractmethod
    def refund(self,transaction_id):
        pass
class RazorpayGateway(PaymentGateway):
    def initiate(self, amount):
        print("Razorpay payment initiated:",amount)
    def verify(self, transaction_id):
        print("Razorpay transaction verified:",transaction_id)
    def refund(self, transaction_id):
        print("Razorpay refund processed:",transaction_id)
class PaypalGateway(PaymentGateway):
    def initiate(self, amount):
        print("Paypal payment initiated:",amount)
    def verify(self, transaction_id):
        print("Paypal transaction verified:",transaction_id)
    def refund(self, transaction_id):
        print("Paypal refund processed:",transaction_id)
r = RazorpayGateway()
r.initiate(500)
r.verify("RZP101")
r.refund("RZP101")
print()
p = PaypalGateway()
p.initiate(1000)
p.verify("PP101")
p.refund("PP101")

# g = PaymentGateway()
# TypeError: Can't instantiate abstract class PaymentGateway without an implementation for abstract methods 'initiate', 'refund', 'verify'
```

```python
# 14

# Abstract Contract Enforcement
# Create an abstract class Animal(ABC) with abstract methods speak() and move(),
# and a concrete method breathe() shared by all. Implement Dog, Bird, and Fish.
# Prove the contract is enforced by trying to instantiate Animal or a subclass that
# forgot to implement speak().
from abc import ABC,abstractmethod
class Aniaml(ABC):
    @abstractmethod
    def speak(self):
        pass
    @abstractmethod
    def move(self):
        pass
    def breathe(self):
        print("Animal is breathing")
class Dog(Aniaml):
    def speak(self):
        print("Dog says woof")
    def move(self):
        print("Dog is walking")
class Bird(Aniaml):
    def speak(self):
        print("Bird says Chrip")
    def move(self):
        print("Bird is flying")
class Fish(Aniaml):
    def speak(self):
        print("Fish makes no sound")
    def move(self):
        print("Fish is swimming")
d = Dog()
d.speak()
d.move()
d.breathe()
print()
b = Bird()
b.speak()
b.move()
b.breathe()
print()
f = Fish()
f.speak()
f.move()
f.breathe()
```

```python
# 15

# Abstract Properties
# Combine abstraction with properties: write an abstract class DataStorage(ABC) with
# an abstract property storage_type and abstract methods save(data), load(key), and
# delete(key). Implement FileStorage and MemoryStorage. Demonstrate that both are
# interchangeable.
from abc import ABC,abstractmethod
class DataStorage(ABC):
    @property
    @abstractmethod
    def storage_type(self):
        pass
    @abstractmethod
    def save(self,data):
        pass
    @abstractmethod
    def load(self,key):
        pass
    @abstractmethod
    def delete(self,key):
        pass
class FileStorage(DataStorage):
    @property
    def storage_type(self):
        return "File Storage"
    def save(self,data):
        print("Saving data to file:",data)
    def load(self,key):
        print("Loading data from file:",key)
    def delete(self,key):
        print("Deleting data from file:",key)
class MemroyStorage(DataStorage):
    @property
    def storage_type(self):
        return "Memory Storage"
    def save(self,data):
        print("Saving data to Memory:",data)
    def load(self,key):
        print("Loading data from file:",key)
    def delete(self,key):
        print("Deleting data from file:",key)
f = FileStorage()
print(f.storage_type)
f.save("Student data")
f.load("Student1")
f.delete("Student1")
print()
m = MemroyStorage()
print(m.storage_type)
f.save("Student data")
f.load("Student1")
f.delete("Student1")
```

```python
# 1. Payment System 
# Create an abstract class Payment with an abstract methodpay(amount)
# Create:
# •	UPI
# •	CreditCard
# •	Cash
# Each class should implement pay() differently.
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def methodpay(self,amount):
        pass
class UPI(Payment):
    def methodpay(self,amount):
        print("Paid Rs.",amount,"using UPI")
class CreditCard(Payment):
    def methodpay(self, amount):
        print("Paid Rs.",amount,"using CreditCard")
class Cash(Payment):
    def methodpay(self, amount):
        print("Paid Rs.",amount,"using Cash")
u = UPI()
u.methodpay(500)
cc = CreditCard()
cc.methodpay(1000)
c = Cash()
c.methodpay(200)
```

```python
# 2. Notification System 
# Create an abstract class Notification with:
# send(message)
# Create:
# •	Email 
# •	SMS 
# •	WhatsApp 
# Each child class should implement send() differently.

from abc import ABC,abstractmethod
class Notification(ABC):
    @abstractmethod
    def send(self,message):
        pass
class Email(Notification):
    def send(self,message):
        print("Sending Email:",message)
class SMS(Notification):
    def send(self,message):
        print("Sending SMS:",message)
class WhatsApp(Notification):
    def send(self,message):
        print("Sending WhataApp Message:",message)
e = Email()
e.send("Your order has been confirmed")
s = SMS()
s.send("Your OTP is 1234")
w = WhatsApp()
w.send("Hello! How are you?")
```

```python
# 3. Login System
# Create an abstract class LoginSystem with:
# login(username, password)
# logout()
# Create:
# •	AdminLogin
# •	UserLogin
# Implement login differently for admin and normal users.
from abc import ABC,abstractmethod
class LoginSystem(ABC):
    @abstractmethod
    def login(self,username,password):
        pass
    @abstractmethod
    def logout(self):
        pass
class AdminLogin(LoginSystem):
    def login(self, username, password):
        if(username=="admin" and password=='admin123'):
            print("Admin Login successful")
        else:
            print("Invalid admin credentials")
    def logout(self):
        print("Admin logged out successfully")
class UserLogin(LoginSystem):
    def login(self,username,password):
        if(username=='user' and password=='user123'):
            print("User Login successful")
        else:
            print("Invaldi user credentials")
    def logout(self):
        print("User logged out successfully")
a = AdminLogin()
a.login("admin","admin123")
a.logout()
print()
u = UserLogin()
u.login("user","user123")
u.logout()
```

```python
# 4. Bank Account
# Create an abstract class BankAccount with two abstract methods:
# deposit(amount)
# withdraw(amount)
# Create:
# •	SavingsAccount
# •	CurrentAccount
# Maintain the balance using an instance variable.

from abc import ABC,abstractmethod
class BankAccount(ABC):
    def __init__(self,balance):
        self.balance = balance
    @abstractmethod
    def deposit(self,amount):
        pass
    @abstractmethod
    def withdraw(self,amount):
        pass
class SavingsAccount(BankAccount):
    def deposit(self, amount):
        self.balance += amount
        print("Savings Account - Deposited",amount)
    def withdraw(self, amount):
        if(amount<=self.balance):
            self.balance -= amount
            print("Savings Account - Withdrawn:",amount)
        else:
            print("Insufficient balance")
class CurrentAccount(BankAccount):
    def deposit(self, amount):
        self.balance += amount
        print("Current Account - Deposited",amount)
    def withdraw(self, amount):
        if(amount<=self.balance):
            self.balance -= amount
            print("Current Account - Withdrawn:",amount)
        else:
            print("Insufficient Balance")
s = SavingsAccount(10000)
s.deposit(2000)
s.withdraw(3000)
print("Savings Balance:",s.balance)
print()
c = CurrentAccount(20000)
s.deposit(5000)
s.withdraw(4000)
print("Current Balance:",c.balance)
```
