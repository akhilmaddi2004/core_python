```python
# what's composition in python

# **Composition** in Python is an object-oriented design principle where a class (**Composite**) contains objects of other classes (**Components**) as attributes, modeling a **"has-a"** relationship. This approach allows the composite class to reuse the implementation of its components without inheriting their interface, promoting **loose coupling** and **flexibility**.

# Key characteristics include:

# *   **Relationship Type**: It represents a **"has-a"** or **"part-of"** relationship, as opposed to inheritance’s **"is-a"** relationship. For example, a `Car` **has an** `Engine`, whereas a `Dog` **is an** `Animal`.
# *   **Code Reuse**: It enables code reuse by combining objects rather than creating rigid class hierarchies, allowing components to be swapped or modified at runtime without affecting the composite class.
# *   **Implementation**: It is achieved by defining a class that includes instances of other classes as instance variables, typically passed via the constructor or defined as attributes.
# *   **Benefits**: Composition reduces dependencies between classes, making the code more modular, easier to test, and adaptable to change compared to inheritance.

# Example:

class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        # Composition: Car has an Engine
        self.engine = Engine() 
    
    def drive(self):
        return self.engine.start() + " -> Car driving"
c1 = Car()
print(c1.drive())
```

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()  # composition: Car "has an" Engine

    def drive(self):
        self.engine.start()     # calling the component's method
        print("Car is driving")

# --- Calling with object ---
car = Car()        # creates a Car (which internally creates an Engine)
car.drive()
```

```python
class Wheel:
    def __init__(self, size):
        self.size = size

class Bike:
    def __init__(self, wheel1, wheel2):
        self.wheel1 = wheel1   # composition via constructor args
        self.wheel2 = wheel2

    def describe(self):
        print(f"Wheels: {self.wheel1.size}, {self.wheel2.size}")

# --- Calling with objects passed in ---
w1 = Wheel(26)
w2 = Wheel(26)
bike = Bike(w1, w2)
bike.describe()   
```

## Inheritance and Composition in Python

```python
# Inheritance and composition are two approaches in Python for building relationships between classes and resusing code.
# Inheritance creates a parent-child realtionship between classes, while composition builds classes by combining objects.

# - Inheritance represents an "is-a" relationship between classes.
# - Composition represents a "has-a" relationship between objects.

# Inheritance:
# It allows a child class to reuse and extend functionality from a parent class.
# Example:

class Vehicle:
    def start(self):
        print("Vehicle started")
class Car(Vehicle):
    def drive(self):
        print("Car is driving")
c = Car()
c.start()
c.drive()

# Explaination:
# Car inherits start() from Vehicle.
# drive() add functionality specific to Car.
# Car is a Vehicle, forming an is-a relationship.
# Child classes can also override inherited methods when different behavior is required.
```

```python
# Composition:
# It combines objects by making one class contain an object of another class.
# - Containing class uses functionality provided by another object.
# - Separate classes remain responsible for their own behavior.
# - Relationship between classes represents has-a rather than is-a.

# Example:
class Engine:
    def start(self):
        print("Engine started")
class Car:
    def __init__(self):
        self.engine = Engine()
    def drive(self):
        self.engine.start()
        print("Car is driving")
car = Car()
car.drive()

# Explaination:
# - Car contains an Engine object through self.engine.
# - drive() delegates engine-related work to that object.
# - Car has an Engine, forming a has-a relationship.
# - Changing the Engine implementation does not require creating a sublcass of Car.
```

```python
# Composition vs Inheritance
# Both approaches support code reuse, but they differ in structure, flexibility and dependency between classes.

# Feature               # Inheritance                                   # Composition
# Relationship          # Is-A                                          # Has-A
# Reuse mechanism       # Inherits class behavior                       # Uses another object
# Dependency            # Child depends on parent hierarchy             # Container depends on contained object
# Behavior change       # Overriding or extending inherited methods     # Delegation or replacing contained objects
# Clas hierarchy        # Creates hierarcy                              # Avoids hierarchy
# Flexibility           # Less flexible for changing realtionships      # More flexible through interchangable objects
# Best suited for       # Closely related class types                   # Combining independent responsibilities

# Example: Inheritance and Composition Together
class Device:
    def power_on(self):
        print("Device is powered on")
class Camera:
    def take_photo(self):
        print("Photo captured")
class Smartphone(Device):
    def __init__(self):
        self.camera = Camera()
    def use_camera(self):
        self.camera.take_photo()
phone = Smartphone()
phone.power_on()
phone.use_camera()

# Explaination:
# - Smartphone inherits power_on() from Device.
# - It contains a Camera object and uses its take_photo() method.
# - Smartphone is a Device, representing inheritance.
# - Camera is a part of Smartphone, representing composition.
```
