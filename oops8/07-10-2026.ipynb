## ABSTRACTION

````python
# Abstraction
# Abstraction in Python is the Object-Oriented Programming(OOP) principle of hiding complex implementation details and exposing only the essential functionality to the user.

# It simplifies interaction with objects by allowing developers to focus on what a method does rather than how it works internally.

# In Python, abstraction is implemented using Abstract Base Classes(ABCs) from the built-in "abc" module.

# An abstract class inherits from "ABC" and cannot be instantiated driectly; it serves as a blueprint for other classes.

# It typically contains one or more abstract methods, which are declared using the @abstractmethod decorator and must be implemented by any subclass.

# Key Components:
# - Abstract Base Class (ABC): A class that cannot be instantiated and defines a common interface for its subclasses.
# - Abstract Method: A method declared in an abstract class without a body(using pass), forcing subclasses to provide a specific implementation.
# - Concrete Method: A standard method within an abstract class that provides a default implementation and can be inherited by subclasses.

# Example Implementation:
from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14*self.radius**2
# shape = Shape() # TypeError: Can't instantiate abstract class
circle = Circle(5)
print(circle.area()) 

# Abstraction promotes code modularity, resuability, and maintainability by enforcing a consistent structure across realted classes while shielding users from internal complexity.
````

````python
# Abstraction:
# Abstraction is a process of hiding implementatino details and exposing only essential features.
# Absstraction tells us what an object can do without telling us how it does it.

# why do we need abstraction:
# 1. Hide complexity
# 2. Support polymorphism
# 3. Provide common structure

# Abstract Class:
# - An abstract class is a class designed as a blueprint for related classes. It may contain abstract methods, normal methods, constructors, and variables.
# - An abstract class containing unimplemented abstract methods can't be instantiated.

# Abstract Method:
# - An abstract method is a method declared in the parent class that requires a concrete sub class to provide an implementation
# - It is marked using @abstractmethod decorator.

# - using super() we call constructor in abstract classes and subclasses.

# Abstract class rules:
# 1. Import ABC and abstractmethod from abc.
# 2. An abstract class can have abstract methods,normal methods, constructors and varialbes.
# 3. An abstract class with unimplemented abstract methods cannot be instantiated.
# 4. A child must implement all inherited abstract methods to become concrete.
# 5. Abstractness can continue through mutlilevel inheritance
# 6. Python does not automatically enforce an exact abstract method signature
# 7. Changing parameter names is harmless; changing required parameter count can break the expected contract.
# 8. super() can call an abstract parent's constructor or normal method
# 9. Abstract classes are useful for enforcing a common design while supporting polymorphism.

# Concrete class vs abstract class:

# - concrete class: object can be created using concrete class
# - abstract class: but can't create object in abstract class

# - Concrete classes are fully implemented classes that can be instantiated directly, while abstract classes serve as incomplete blueprints that cannot be instantiated and enforce a contract on their subclasses.

# - In Python, abstract classes are created by inheriting from abc.ABC and using the @abstractmethod decorator to declare methods that subclasses must to override. Key distinctions include:

# - Instantiation: Concrete classes can be instantiated immediately; abstract classes raise a TypeError if attempted.
# - Implementation: Concrete classes provide full method bodies. Abstract classes may contain both abstract methods(no implementation) and concrete methods(shared implementation).
# - Subclassing: A subclass remains abstract if it fails to implement all inherited abstract methods. It becomes concrete only when every abstract method is overridden.
# - Purpose: Abstract classes enforce a consistent interface and promote code reuse through shared concrete methods, whereas concrete classes represent specific, usable entities.
````

````python
# ### Abstract Class — 9 Easy Rules

# **1. Import first**
# Before creating an abstract class, import:
# ```python
# from abc import ABC, abstractmethod
# ```

# **2. Abstract class can contain different things**
# An abstract class can have:
# - Abstract methods
# - Normal methods
# - Constructor (`__init__`)
# - Variables

# So, an abstract class does **not** mean every method must be abstract.

# **3. Cannot create an object if abstract methods are unfinished**
# If an abstract class has an abstract method that is not implemented, we **cannot create its object**.

# ```python
# obj = Animal()   # Error
# ```

# **4. Child must complete all abstract methods**
# If a child class inherits an abstract class, the child must implement **all abstract methods** of the parent.

# Only then can we create an object of the child class.

# **5. Abstractness can continue in multilevel inheritance**
# If the first child does not implement all abstract methods, it also remains **abstract**.

# The next child can implement the remaining methods and become concrete.

# ```text
# Parent (Abstract)
#        ↓
# Child  (Still Abstract)
#        ↓
# Grandchild (Concrete)
# ```

# **6. Python does not automatically check the exact method signature**
# Python checks whether the abstract method is implemented, but it does **not strictly force the child to use exactly the same parameters**.

# For example, the parent may have:

# ```python
# def show(self, name):
#     pass
# ```

# The child can technically write:

# ```python
# def show(self):
#     print("Hello")
# ```

# Python can consider the abstract method implemented, even though the signature is different.

# **7. Parameter names can change, but required parameter count matters**
# Changing the **parameter name** is normally okay:

# ```python
# def show(self, name):
# ```

# can become:

# ```python
# def show(self, student_name):
# ```

# But changing the **number of required parameters** can cause problems when the method is called according to the parent's expected design.

# Easy memory:

# > **Parameter name can change; parameter count should follow the expected contract.**

# **8. `super()` can call the abstract parent's constructor or normal method**
# Even if the parent class is abstract, `super()` can still be used to call its:

# - Constructor
# - Normal methods

# For example:

# ```python
# super().__init__()
# ```

# or:

# ```python
# super().display()
# ```

# An abstract method itself is generally meant to be implemented by the child rather than used as the implementation.

# **9. Main purpose of abstract classes**
# Abstract classes are used when we want different child classes to follow a **common design**.

# They make sure certain methods exist in the child classes while still allowing each child to provide its **own implementation**.

# This works well with **polymorphism**.

# ### 🧠 Super-easy memory version

# Remember these 9 points:

# 1. **Import** → `ABC`, `abstractmethod`
# 2. **Can contain** → abstract + normal + constructor + variables
# 3. **Incomplete abstract class** → no object
# 4. **Child** → implement all abstract methods
# 5. **Not implemented** → abstractness continues
# 6. **Exact signature** → Python doesn't automatically enforce it
# 7. **Parameter name** → can change; **parameter count** → should follow the contract
# 8. **`super()`** → can call parent's constructor/normal method
# 9. **Purpose** → common design + polymorphism

# **One-line idea:**

# > **Abstract class = Parent gives the rules; child classes must follow those rules and can provide their own implementations.**
````

````python
# Q) create a python application with an abstract class upload with an abstract method upload_content, implement three subclasses photoUpload, videoUpload and storyUpload
from abc import ABC,abstractmethod
class Upload(ABC):
    @abstractmethod
    def upload_content(self):
        pass
class photoUpload(Upload):
    def upload_content(self):
        print("Uploading the photo")
        print("Encrypting the photo")
        print("Decrypting the photo")
        print("Uploaded the photo")
class videoUpload(Upload):
    def upload_content(self):
        print("Uploading the video")
        print("Encrypting the video")
        print("Decrypting the video")
        print("Uploaded the video")
class storyUpload(Upload):
    def upload_content(self):
        print("Uploading the story")
        print("Encrypting the story")
        print("Decrypting the story")
        print("Uploaded the story")
p1 = photoUpload()
p1.upload_content()
print()
v1 = videoUpload()
v1.upload_content()
print()
s1 = storyUpload()
s1.upload_content()

# Abstract class = gives the rule.
# Child class = follows the rule and gives the implementation.
````
