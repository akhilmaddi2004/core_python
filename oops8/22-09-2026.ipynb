```python
# create a class A that has two class variables and two instance variables.
# create another class B that is child class of A which has one instance & one class variable access all these properties using object b
class A:
    c1 = "hi"
    c2 = 123
    def __init__(self,x,y):
        self.x = x
        self.y = y
class B(A):
    c3 = True
    def __init__(self,z):
        self.z = z
        A.__init__(self,20,30)
b = B(10)
print(b.x)
print(b.y)
print(b.z)
print(A.c1)
print(A.c2)
print(B.c3)
```

```python
# super(): It will pass current object address to child class.
# - super() returns a proxy object that delegates method calls to the next class in MRO.
# - It lets you call a parent class property without specifying the name of the parent.
# - MRO - method resolution order.
class A:
    c1 = "hi"
    c2 = "True"
    def __init__(self,x,y):
        self.x = x
        self.y = y 
class B(A):
    c3 = True
    def __init__(self,z):
        self.z = z
        super().__init__(20,30)
b = B(10)
print("x:",b.x)
print("y:",b.y)
print("z:",b.z)
print("c1:",A.c1)
print("c2",A.c2)
print("c3:",B.c3)
print()
class A:
    c1 = "hey"
    c2 = 123
    def __init__(self,x,y):
        self.x = x
        self.y = y
class B(A):
    c3 = True
    def __init__(self, z, x, y):
        self.z = z
        super().__init__(x, y)
b = B(10,20,30)
print("x:",b.x)
print("y:",b.y)
print("z:",b.z)
print("c1:",A.c1)
print("c2",A.c2)
print("c3:",B.c3)
```

```python
# multi-level inheritance
class A:
    def __init__(self,x):
        self.x = x
class B(A):
    def __init__(self,x,y):
        self.y = y
        super().__init__(x)
class C(B):
    def __init__(self,x,y,z):
        self.z = z
        super().__init__(x,y)
c = C(10,20,30)
print("x:",c.x)
print("y:",c.y)
print("z:",c.z)
```
