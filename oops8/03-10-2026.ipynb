```python
# 5 Method Overriding
# Build a payment system with a base class Payment(amount) and three subclasses:
# CreditCard, UPI, and NetBanking. Each overrides a method process() with its own
# logic. Write a function checkout(payment) that calls process() on any payment object
# and demonstrate polymorphism.
class Payment:
    def __init__(self,balance):
        self.balance = balance
        print("Object Created!! Your current Balance is Rs.",balance)
    def process(self,amount):
        print("Processing Payment!")
        self.balance += amount
        print("Payment Processed! Your current balance is Rs.",self.balance)
class CreditCard(Payment):
    def process(self, amount):
        print("Processing Payment of Rs.",amount,"via CreditCard")
        self.balance += amount
        print("Payment Processsed! Your current balance is Rs.",self.balance)
class UPI(Payment):
    def process(self,amount):
        print("Processing Payment of Rs.",amount,"via UPI")
        self.balance += amount
        print("Payment Processed! Your current Balance is Rs.",self.balance)
class NetBanking(Payment):
    def process(self,amount):
        print("Processing Payment of Rs.",amount,"via NetBanking")
        self.balance += amount
        print("Payment Processed! Your current Balance is Rs.",self.balance)
def checkout(payment):
    payment.process(1200)
checkout(CreditCard(1500))
print()
checkout(UPI(1200))
print()
checkout(NetBanking(1500))
```

```python
# 6 Duck Typing
# Create three completely unrelated classes: PDFReport, ExcelSheet, and
# EmailMessage. Each has a method send(). Write a function dispatch(item) that calls
# send() using duck typing. Prove that no inheritance is needed.
class PDFReport:
    def send(self):
        print("Sending PDF...")
class ExcelSheet:
    def send(self):
        print("Sending ExcelSheet...")
class EmailMessage:
    def send(self):
        print("Sending Email Message...")
def dispatch(item):
    item.send()
dispatch(PDFReport())
dispatch(ExcelSheet())
dispatch(EmailMessage())
```

```python
# 7 Polymorphism + Duck Typing
# Write a function total_area(shapes) that takes a list of any shape objects and returns
# their combined area. Use duck typing - any object with an area() method should
# work. Test with Circle, Rectangle, Triangle, and a custom class Hexagon.
import math
class Circle:
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius**2
class Rectangle:
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth
    def area(self):
        return self.length * self.breadth
class Triangle:
    def __init__(self,base,height):
        self.base = base
        self.height = height
    def area(self):
        return 0.5 * self.base * self.height
class Hexagon:
    def __init__(self,side):
        self.side = side
    def area(self):
        return (3*math.sqrt(3)/2) * self.side **2
def total_area(shapes):
    area = 0
    for shape in shapes:
        area += shape.area()
    print(area)
shapes = [Circle(5),Rectangle(10,2),Triangle(5,2),Hexagon(3)]
total_area(shapes)
```

```python
# 8 Concept Question
# Explain the difference between method overriding and method overloading. Python
# does not natively support overloading - show how to simulate it using *args or
# default parameters.

# | Method Overriding                            | Method Overloading                                                       |
# | -------------------------------------------- | ------------------------------------------------------------------------ |
# | Happens between **parent and child classes** | Usually multiple methods with the **same name but different parameters** |
# | Child provides a new implementation          | Same method handles different numbers/types of arguments                 |
# | Python supports it directly                  | Python does **not natively support** it                                  |
# | Used in inheritance                          | Can be simulated using `*args` or default arguments                      |

# 1. Method overriding
class Animal:
    def sound(self):
        print("Animal Sound")
class Dog(Animal):
    def sound(self):
        print("Dog Barks")
d = Dog()
d.sound()

# 2. Method overloading
class Calculator:
    def add(self,a,b):
        print(a+b)
    def add(self,a,b,c):
        print(a+b+c)
c = Calculator()
c.add(1,2,3)

class Calculator:
    def add(self,*args):
        if len(args)==1:
            return args[0]
        elif len(args) == 2:
            return args[0] + args[1]
        else:
            return "Invalid number of arguments"
c1 = Calculator()
print(c1.add(8))
print(c1.add(8,7))

class Calculator:
    def add(self,*args):
        total = 0
        for i in args:
            total += i
        return total
c2 = Calculator()
print(c2.add(1,2,3,42,3,1))

class Calculator:
    def add(self,a,b=0):
        return a+b
c3 = Calculator()
print(c3.add(9))
print(c3.add(10,20))
```
