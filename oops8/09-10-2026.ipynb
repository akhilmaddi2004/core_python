```python
# 14

# Abstract Contract Enforcement
# Create an abstract class Animal(ABC) with abstract methods speak() and move(),
# and a concrete method breathe() shared by all. Implement Dog, Bird, and Fish.
# Prove the contract is enforced by trying to instantiate Animal or a subclass that
# forgot to implement speak().
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass
    @abstractmethod
    def move(self):
        pass
    def breathe(self):
        pass
class Dog(Animal):
    def speak(self):
        print("Dog braks")
    def move(self):
        print("Dog Runs with 4 limbs")
class Bird(Animal):
    def speak(self):
        print("Bird chirps")
    def move(self):
        print("Bird flies")
class Fish(Animal):
    def speak(self):
        print("Whispers..")
    def move(self):
        print("Fish swims")
class Cat(Animal):
    def move(self):
        print("Cat walks")
# c = Cat()
# TypeError: Can't instantiate abstract class Cat without an implementation for abstract method 'speak'
```

```python
# 15

# Abstract Properties
# Combine abstraction with properties: write an abstract class DataStorage(ABC) with
# an abstract property storage_type and abstract methods save(data), load(key), and
# delete(key). Implement FileStorage and MemoryStorage. Demonstrate that both are
# interchangeable.
from abc import abstractmethod,ABC
class DataStorage(ABC):
    @property
    @abstractmethod
    def storage_type(self):
        pass
    @abstractmethod
    def save(self,data):
        pass
    @abstractmethod
    def load(self,key):
        pass
    @abstractmethod
    def delete(self,key):
        pass
class FileStorage(DataStorage):
    @property
    def storage_type(self):
        return FileStorage.__name__
    def save(self,data):
        print(f"Saving data: {data}")
    def load(self,key):
        print(f"Loading data: {key}")
    def delete(self, key):
        print(f"Deleting data: {key}")
class MemoryStorage(DataStorage):
    @property
    def storage_type(self):
        return MemoryStorage.__name__
    def save(self,data):
        print(f"Saving data: {data}")
    def load(self,key):
        print(f"Loading data: {key}")
    def delete(self,key):
        print(f"Deleting data: {key}")
file_obj = FileStorage()
file_obj.load("Hi!!")
memory_obj = MemoryStorage()
memory_obj.load("Hey!!")
print(file_obj.storage_type)
```

### Questions On @property and @property.setter

```python
# 1. Create an Employee class with a private __salary. Use @property to get the salary and @salary.setter to modify it. Create an employee object and update the salary.
class Employee:
    def __init__(self,salary):
        self.__salary = salary
    @property
    def salary(self):
        return self.__salary
    @salary.setter
    def salary(self,new_salary):
        self.__salary = new_salary
e = Employee(10000)
print("old salary:",e.salary)
e.salary = 20000
print("new salary",e.salary)
```

```python
# 2. Create a BankAccount class with a private __balance. Implement balance using @property and @balance.setter. Add deposit(amount) and withdraw(amount) methods that modify the balance through the property.
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance
    @property
    def balance(self):
        return self.__balance
    @balance.setter
    def balance(self,new_balance):
        self.__balance = new_balance
    def deposit(self,amount):
        if(amount>0):
            self.__balance += amount
            print("Amount Depoisted successfully")
        else:
            print("Invalid deposit amount")
    def withdraw(self,amount):
        if(amount<=self.__balance):
            self.__balance -= amount
            print("Amount Withdraw successfully")
        else:
            print("INsufficient balance")
b1 = BankAccount(1000)
b1.deposit(500)
b1.withdraw(200)
print("Balance:",b1.balance)
```

```python
# 3. Create a Person class with private __first_name and __last_name. Create properties for both. Create a read-only property full_name that returns:first_name + " " + last_name
class Person:
    def __init__(self,first_name,last_name):
        self.__first_name = first_name
        self.__last_name = last_name
    @property
    def first_name(self):
        return self.__first_name
    @first_name.setter
    def first_name(self,name):
        self.__first_name = name
    @property
    def last_name(self):
        return self.__last_name
    @last_name.setter
    def last_name(self,name):
        self.__last_name = name
    @property
    def full_name(self):
        return self.__first_name+" "+self.__last_name

p1 = Person("Akhil","Maddi")
print("First Name:",p1.first_name)
print("Last Name:",p1.last_name)
print("Full Name:",p1.full_name)
p1.first_name = "Arjun"
print("Updated fullname:",p1.full_name)
```

```python
# 4. Create a ShoppingCart class with a private __total. Create a property total with getter and setter. Add methods add_item(price) and remove_item(price) that modify the total using the property.
class ShoppingCart:
    def __init__(self):
        self.__total = 0
    @property
    def total(self):
        return self.__total
    @total.setter
    def total(self,new_total):
        self.__total = new_total
    def add_item(self,price):
        if(price>0):
            self.__total += price
    def remove_item(self,price):
        if(price>0 and price<=self.__total):
            self.__total -= price
        else:
            print("Invalid Price")
c1 = ShoppingCart()
print(c1.total)
c1.add_item(100)
c1.add_item(200)
print("Total:",c1.total)
c1.remove_item(100)
print("Total after removal:",c1.total)
```

```python
# 5.Create a Student class with private variables __name and __marks. Create properties name and marks with getter and setter methods. Create a read-only property result that returns "Pass" if marks are 40 or above, otherwise "Fail".
class Student:
    def __init__(self,name,marks):
        self.__name = name
        self.__marks = marks
    @property
    def name(self):
        return self.__name
    @name.setter
    def name(self,name):
        self.__name = name
    @property
    def marks(self):
        return self.__marks
    @marks.setter
    def marks(self,new_marks):
        self.__marks = new_marks
    @property
    def result(self):
        if(self.__marks>=40):
            return "Pass"
        else:
            return "Fail"
s1 = Student("Akhil",80)
print("Name:",s1.name)
print("Marks:",s1.marks)
print("Result:",s1.result)
s1.name = "rahul"
s1.marks = 35
print("Name:",s1.name)
print("Marks:",s1.marks)
print("Result:",s1.result)
```
