## Polymorphism

```python
# polymorphism in python

# **Polymorphism** in Python allows objects of different classes to be treated uniformly through a common interface, enabling the same method or operator to behave differently based on the object's type. This concept is central to **Object-Oriented Programming (OOP)** and facilitates code reusability, flexibility, and maintainability by allowing developers to write generic code that works with various data types.

# Python implements polymorphism primarily through three mechanisms:

# *   **Duck Typing**: A form of dynamic polymorphism where an object's suitability is determined by the presence of specific methods or properties rather than its actual class. If an object supports a method, it can be used interchangeably with any other object supporting that same method, regardless of inheritance relationships.
# *   **Method Overriding**: A form of runtime polymorphism achieved through inheritance, where a subclass provides a specific implementation for a method already defined in its parent class. When the method is called, Python executes the subclass's version based on the actual object instance.
# *   **Operator Overloading**: Built-in operators like `+` or `len()` exhibit polymorphic behavior by performing different actions depending on the operands' types (e.g., addition for numbers vs. concatenation for strings).

# Key benefits include **code reusability**, as a single function can process multiple object types, and **abstraction**, which hides complex implementation details behind a simple, unified interface.
```

```python
# In Python, polymorphism means the same function, operation, method cn behave differently based on the object or data it is working with.

class A:
    def m1(self):
        print("m1-A")
class B:
    def m1(self):
        print("m1-B")
class C:
    def m1(self):
        print("m1-C")
a = A()
b = B()
c = C()
a.m1()
b.m1()
c.m1()
```

```python
# method overriding
# method overriding happens when a child class provides it's own implementation of a method that already exists in it's parent class.
# The child's version replaces the parent's version when called on a child object.
class Bank:
    def deposit(self,amount):
        print("Deposited Rs.",amount)
class Gpay(Bank):
    def deposit(self, amount):
        print("Deposted Rs.:",amount)
        print("You've gotten Rs.",(amount*0.2),"cashback")
class Phpay(Bank):
    def deposit(self,amount):
        print("Depostied Rs.:",amount)
        print("You've gotten Rs.",(amount*0.1),"cashback")
        print("You've gotten 1 scartch card")
obj1 = Gpay()
obj1.deposit(500)
obj2 = Phpay()
obj2.deposit(500)

# method overloading doesn't exist in python like Java, because we are using *args in method to pass unknown no. of parameters.
# method overloading means same method in different classess with different no. of parameters.
# But we have operator overloading in python like dunder methods
```

```python
# Types of polymorphism in python:
# In general polymorphism of two types:
# 1. overloading 
# 2. overriding

# 1. Method overloading:
# - Method overloading is not possible in python
class A:
    def add(self,a,b):
        print(a+b)
    def add(self,a,b,c):
        print(a+b+c)
# in class A => the def add(a,b)
# first stores print(a+b)
# then when we create def add(a,b,c)
# it stores only print(a+b+c) and first stored will be removed permenantly

# A().add(10,20) #TypeError: A.add() missing 1 required positional argument: 'c'
print("Method overloading:")
print("add")
A().add(10,20,30)

# Then when we call add(a,b) we get error with only 2 parameters
# Python doesn't support traditional method overloading where multiple methods have the same name in same class.
# In the above example the second add() method replaces the first one and if you call 
# A().add(10,20) #TypeError: A.add() missing 1 required positional argument: 'c'

# we can write like below but in case if we don't know something how many parameters , use default values
class A:
    def add(self,a,b,c=0):
        print(a+b+c)
print("Default")
A().add(10,20)
print()

# Operator polymorphism:
# '+' operator has different functionality based on object type.
print("Operator polymorphism")
print(10+20) # produces an integer object 30
print("A"+"B") # as result of concatenation
# using "+" operator for 2 lists joins the 2 lists.
print([1,2]+[3,4])

# Function Polymorphism:
# For example, len("apple")
print("length of string 'apple':",len('apple'))
print("length of a list:",len([10,20,30]))
print("length of dictionary:",len({'a':[10,20],'b':30,'c':[40,50]}))
```

```python
# Ducktyping 
# The concept of ducktyping comes from the phrase :
# "If it walks like a duck and quacks like a duck then it must be a duck".

# Ducktyping is python's approach to polymorphism without inheritance
# If an object has methods and attributes you need, you can use it regarless of its class.
# you don't check is a relationship(inhertance) you just call the method if it works great else you get an attribute error.

class A:
    def m1(self):
        print("m1-A")
class B:
    def m1(self):
        print("m1-B")
class C:
    def m1(self):
        print("m1-C")
a = A()
b = B()
c = C()
a.m1()
b.m1()
c.m1()
print()
l = [a,b,c]
for obj in l:
    obj.m1()
```
