```python
# 3. Create: 
# • Abstract class PaymentMethod with pay(), validate() 
# • Subclasses: CardPayment, WalletPayment, UPIPayment 
# • Encapsulate user balance 
# • Use @property to control reading available funds 
# • Overload + operator to combine two payment methods into “split payment” 
# • Demonstrate polymorphism through a checkout loop. 
from abc import ABC,abstractmethod
class PaymentMethod(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
    @abstractmethod
    def validate(self,amount):
        return amount>0
    # Overload + operator
    def __add__(self, other):
        return SplitPayment(self,other)
class CardPayment(PaymentMethod):
    def validate(self,amount):
        return amount>0
    def pay(self,amount):
        if(self.validate(amount)):
            print("Paid Rs.",amount,"using Card")
            return True
        print("Invalid card payment amount")
        return False
class WalletPayment(PaymentMethod):
    def __init__(self,balance):
        self.__balance = balance
    @property
    def avaliable_funds(self):
        return self.__balance
    def validate(self, amount):
        return amount>0 and amount <= self.__balance
    def pay(self,amount):
        if(self.validate(amount)):
            print("Paid Rs.",amount,"using Wallet")
            return True
        print("Invalid amount or insufficient wallet funds.")
        return False
class UPIPayment(PaymentMethod):
    def validate(self, amount):
        return amount>0
    def pay(self,amount):
        if(self.validate(amount)):
            print("Paid Rs.",amount,"using UPI")
            return True
        print("Invalid UPI payment amount")
        return False
# Split Payment
class SplitPayment(PaymentMethod):
    def __init__(self,payment1,payment2):
        self.payment1 = payment1
        self.payment2 = payment2
    def validate(self, amount):
        if(amount<=0):
            return False
        half = amount/2
        return (self.payment1.validate(half)
                and self.payment2.validate(half))
    def pay(self,amount):
        if(self.validate(amount)):
            half = amount/2
            self.payment1.pay(half)
            self.payment2.pay(half)
            print("Split payment completed")
            return True
        print("Split Payment validation failed")
        return False
# creating payment objects
card = CardPayment()
wallet = WalletPayment(1000)
upi = UPIPayment()

# combine two payment methods using +
split_payment = card + wallet

# checkout loop: polymorphism
payments = [card,wallet,upi,split_payment]
amounts = [500,300,200,400]
for payment,amount in zip(payments,amounts):
    print("\n----checkouts----")
    payment.pay(amount)
# read wallet balance through property
print("\nWallet Avaliable Funds:",wallet.avaliable_funds)
```

```python
# 6. Create: 
# • Product class with private price and quantity 
# • Warehouse class containing multiple products 
# • Overload: 
# o + to merge warehouses 
# o len() to return number of unique products 
# o in operator to check if product exists 
# • Provide class method to track total warehouses created 

# product class
class Product:
    def __init__(self,name,price,quantity):
        self.name = name
        self.__price = price
        self.__quantity = quantity
    @property
    def price(self):
        return self.__price
    @property
    def quantity(self):
        return self.__quantity
    def add_quantity(self,quantity):
        if(quantity>0):
            self.__quantity += quantity
    def __str__(self):
        return (
            f"Product: {self.name},"
            f"Price: Rs.{self.price},"
            f"Quantity: {self.quantity}"
        )
# Warehouse class
class Warehouse:
    total_warehouses = 0
    def __init__(self,products = None):
        self.products = {}
        if(products is not None):
            for product in products:
                self.add_product(product)
        Warehouse.total_warehouses += 1
    def add_product(self,product):
        if(product.name in self.products):
            exisiting_product = self.products[product.name]
            if(exisiting_product.price != product.price):
                print("Cannot merge products with different prices")
                return 
            exisiting_product.add_quantity(product.quantity)
        else:
            self.products[product.name] = product
    # overload + operator
    def __add__(self, other):
        if not isinstance(other,Warehouse):
            return NotImplemented
        merged_products = []
        for product in self.products.values():
            merged_products.append(Product(product.name,product.price,product.quantity))
        for product in other.products.values():
            merged_products.append(Product(product.name,product.price,product.quantity))
        return Warehouse(merged_products)
    # Overload len()
    def __len__(self):
        return len(self.products)
    # overload in operator
    def __contains__(self, product_name):
        return product_name in self.products
    # class method
    @classmethod
    def get_total_warehouses(cls):
        return cls.total_warehouses
    def display(self):
        for product in self.products.values():
            print(product)
# creating products
p1 = Product("Laptop",50000,5)
p2 = Product("Mouse",500,10)

p3 = Product("Laptop",50000,3)
p4 = Product("Keyboard",1000,4)

# creating warehouses
w1 = Warehouse([p1,p2])
w2 = Warehouse([p3,p4])

print("----Warehouse 1----")
w1.display()
print("----Warehouse 2----")
w2.display()
# merge warehouses using +
w3 = w1 + w2 
print("\n----Merged Warehouse ----")
w3.display()
# check number of unique products
print("\nUnique Products:",len(w3))
# check whether a product exists
print("Laptop" in w3)
print("Phone" in w3)
# Count warehouses
print("Total Warehouses:",Warehouse.get_total_warehouses())
```

```python
# 7. Design: 
# • Abstract class MediaFile with play(), stop() 
# • Subclasses: MP3File, MP4File, WAVFile 
# • Private file path validation done internally 
# • A function start_player(media) that works with ANY object that has play() 
# (duck typing) 
# Demonstrate mixing true polymorphism + duck typing. 
from abc import abstractmethod,ABC
class MediaFile(ABC):
    def __init__(self,file_path):
        self.__file_path = file_path
    @property
    def file_path(self):
        return self.__file_path
    @abstractmethod
    def play(self):
        pass
    @abstractmethod
    def stop(self):
        pass
class MP3File(MediaFile):
    def __init__(self, file_path):
        if(not file_path.endswith(".mp3")):
            raise ValueError("Invalid MP3 file path")
        super().__init__(file_path)
    def play(self):
        print("Playing MP3:",self.file_path)
    def stop(self):
        print("Stopping MP3")
class MP4File(MediaFile):
    def __init__(self, file_path):
        if(not file_path.endswith(".mp4")):
            raise ValueError("Invalid MP4 file path")
        super().__init__(file_path)
    def play(self):
        print("Playing MP4:",self.file_path)
    def stop(self):
        print("Stopping MP4 video")
class WAVFile(MediaFile):
    def __init__(self, file_path):
        if(not file_path.endswith(".wav")):
            raise ValueError("Invalid WAV file path")
        super().__init__(file_path)
    def play(self):
        print("Playing WAV:",self.file_path)
    def stop(self):
        print("Stopping WAV")
# Ducktyping function
def start_player(media):
    media.play()
# an unrelated class: does not inherit from MediaFile
class OnlineRadio:
    def play(self):
        print("Playing online radio")
# True polymorphism
m1 = MP3File("songs.mp3")
m2 = MP4File("movie.mp4")
m3 = WAVFile("sound.wav")
m1.play()
m2.play()
m3.play()
m1.stop()
m2.stop()
m3.stop()
print("\nDuck typing:")
start_player(m1)
start_player(m2)
start_player(OnlineRadio())
```

```python
# 8. Create: 
# • Abstract class StatementFormatter 
# • Subclasses: PDFFormatter, JSONFormatter, TextFormatter 
# • Overload __call__() so that formatters can be used like functions 
# • Overload __repr__ for debugging 
# • Demonstrate polymorphic behavior in a reporting pipeline 
from abc import ABC,abstractmethod
import json
class StatementFormatter(ABC):
    @abstractmethod
    def __call__(self,statement):
        pass
    def __repr__(self):
        return self.__class__.__name__ + "()"
class PDFFormatter(StatementFormatter):
    def __call__(self, statement):
        return "PDF Statement:" + str(statement)
class JSONFormatter(StatementFormatter):
    def __call__(self, statement):
        return json.dumps(statement)
class TextForamtter(StatementFormatter):
    def __call__(self, statement):
        return str(statement)
# statement data
statement = {
    "name":"AKhil",
    "balance":5000
}
# create formatters
formatters = [
    PDFFormatter(),
    JSONFormatter(),
    TextForamtter()
]
# reporting pipeline
for fromatter in formatters:
    print(repr(fromatter))
    print(fromatter(statement))
    print()
```

```python
# 9. Classes: 
# • LightDevice 
# • SecurityDevice 
# • SmartCamera(LightDevice, SecurityDevice) 
# Requirements: 
# • Resolve method conflicts using MRO 
# • Encapsulate internal camera logs 
# • SmartCamera overrides both parents’ behaviors 
# • Use super() responsibly in multiple inheritance
class LightDevice:
    def status(self):
        print("Light device is working")
    def turn_on(self):
        print("Light is ON")
class SecurityDevice:
    def status(self):
        print("Security device is working")
    def activate_security(self):
        print("Security system activated")
class SmartCamera(LightDevice,SecurityDevice):
    def __init__(self):
        self.__logs = []
    def status(self):
        super().status()
        print("Smaart camera is working")
    def turn_on(self):
        super().turn_on()
        print("Camera light and recording are ON")
        self.__logs.append("Camera Turned on")
    def activate_security(self):
        super().activate_security()
        print("Camera monitoring is active")
        self.__logs.append("Security activated")
    @property
    def logs(self):
        return self.__logs.copy()
camera = SmartCamera()
# Method conflict resolved using MRO
camera.status()
# Overridden parent behaviours
camera.turn_on()
camera.activate_security()
# private logs
print("Camera logs:",camera.logs)
# Check MRO
print("MRO:",SmartCamera.mro())
```

```python
# 10. Create: 
# • Abstract class MenuItem with get_price() 
# • Subclasses: Pizza, Burger, Drink 
# • Order class containing a list of items (composition) 
# • Encapsulate the list internally 
# • Override methods to apply custom pricing logic for each food type 
from abc import ABC,abstractmethod
class MenuItem(ABC):
    @abstractmethod
    def get_price(self):
        pass

class Pizza(MenuItem):
    def __init__(self,price,extra_toppings):
        self.price = price
        self.extra_toppings = extra_toppings
    def get_price(self):
        return self.price + self.extra_toppings * 30
class Burger(MenuItem):
    def __init__(self,price,extra_cheese):
        self.price = price
        self.extra_cheese = extra_cheese
    def get_price(self):
        return self.price + self.extra_cheese * 20
class Drink(MenuItem):
    def __init__(self,size):
        self.size = size
    def get_price(self):
        if(self.size=="small"):
            return 30 
        elif(self.size=="medium"):
            return 50
        else:
            return 70

class Order:
    def __init__(self):
        self.__items = []
    def add_items(self,item):
        self.__items.append(item)
    def show_order(self):
        total = 0
        for item in self.__items:
            price = item.get_price()
            print(type(item).__name__,":",price)
            total += price
        print("Total bill:",total)
# create food items
p1 = Pizza(200,2)
b1 = Burger(100,1)
d1 = Drink("medium")
# create order
order = Order()
order.add_items(p1)
order.add_items(b1)
order.add_items(d1)
# display bill
order.show_order()
```

```python
# 11. Create: 
# • Class Applicant with private skills list 
# • Overload: 
# o + to add skill 
# o - to remove skill 
# o == to compare applicants who have identical skill sets 
# • Use inheritance to create ExperiencedApplicant with additional fields 
class Applicant:
    def __init__(self,name,skills):
        self.name = name
        self.__skills = skills.copy()
    def __add__(self, skill):
        if(skill not in self.__skills):
           self.__skills.append(skill)
        return self
    def __sub__(self, skill):
        if skill in self.__skills:
            self.__skills.remove(skill)
        return self
    def __eq__(self, other):
        return set(self.__skills) == set(other.__skills)
    def display(self):
        print("Name:",self.name)
        print("Skills:",self.__skills)

class ExperiencedApplicant(Applicant):
    def __init__(self, name, skills,experience,company):
        super().__init__(name, skills)
        self.experience = experience
        self.company = company
    def display(self):
        super().display()
        print("Experience:",self.experience,"years")
        print("Previous Company:",self.company)
# create applicants
a1 = Applicant("Akhil",["Python","SQL"])
a2 = Applicant("Ravi",["SQL","Python"])
# add a skill
a1 + "Django"
a1.display()
# remove a skill
a1 - "SQL"
print("Applicant 1")
a1.display()
print()
print("Applicant 2")
a2.display()
print()
# compare applicants
print("Same skills:",a1 == a2)
# Inheritance
a3 = ExperiencedApplicant("Arjun",["Python","SQL"],3,"ABC Company")
print()
a3.display()
```

```python
# 12. Create: 
# • Character → base class 
# • Warrior, Archer, Mage subclasses 
# Each subclass: 
# • Overrides attack() 
# • Encapsulates health with @property 
# • Prevents negative HP 
# • Uses class attributes for shared attributes (e.g., stamina_cost) 
# Demonstrate polymorphic combat simulation.

class Character:
    stamina_cost = 10
    def __init__(self,name,health):
        self.name = name
        self.__health = 0
        self.health = health
    @property
    def health(self):
        return self.__health
    @health.setter
    def health(self,value):
        if(value<0):
            self.__health = 0
        else:
            self.__health = value
    def attack(self):
        pass

class Warrior(Character):
    stamina_cost = 15
    def attack(self):
        print(self.name,"attacks with a sword!")
class Archer(Character):
    stamina_cost = 10
    def attack(self):
        print(self.name,"attacks with a bow!")
class Mage(Character):
    stamina_cost = 20
    def attack(self):
        print(self.name,"casts a magic spell!")
# create characters
characters = [
    Warrior("Thor",100),
    Archer("Robin",80),
    Mage("Merlin",60)
]
# Polymorphism combat simulation
for character in characters:
    character.attack()
    print("Health:",character.health)
    print("Stamina cost:",character.stamina_cost)
    print()
# test negative health
characters[0].health = -20
print("Warrior health after negative update:",characters[0].health)
```

```python
# 13. Build: 
# • Transport abstract class 
# • Subclasses: Taxi, Bus, Train 
# • Each implements: 
# o calculate_fare() differently 
# • Use static method to validate distance 
# • Encapsulate fare state 
# • Add class method to update government tax slab 
from abc import ABC,abstractmethod
class Transport(ABC):
    tax_rate = 5 # govt tax percentage
    def __init__(self,distance):
        if not Transport.validate_distance(distance):
            raise ValueError("Distance must be greater than zero")
        self.distance = distance
        self.__fare = 0
    @staticmethod
    def validate_distance(distance):
        return distance>0
    @abstractmethod
    def calculate_fare(self):
        pass
    @property
    def fare(self):
        return self.__fare
    def set_fare(self,amount):
        self.__fare = amount
    @classmethod
    def update_tax(cls,new_tax):
        if(new_tax>=0):
            Transport.tax_rate = new_tax

class Taxi(Transport):
    def calculate_fare(self):
        amount = self.distance * 20
        amount = amount + amount * Transport.tax_rate/100
        self.set_fare(amount)
        return self.fare
class Bus(Transport):
    def calculate_fare(self):
        amount = self.distance * 5
        amount = amount + amount * Transport.tax_rate/100
        self.set_fare(amount)
        return self.fare
class Train(Transport):
    def calculate_fare(self):
        amount = self.distance * 3
        amount = amount + amount * Transport.tax_rate/100
        self.set_fare(amount)
        return self.fare
# create transport objects
transports =[
    Taxi(10),
    Bus(10),
    Train(10)
]
# polymorphism
for transport in transports:
    print(type(transport).__name__,"Fare:",transport.calculate_fare())
# update government tax
Transport.update_tax(10)
print("\nAfter tax update:")
for transport in transports:
    print(type(transport).__name__,"Fare:",transport.calculate_fare())
```

```python
# 14. Design: 
# • Abstract class Model with train(), predict() 
# • Implement LinearRegressionModel and DecisionTreeModel(just print or 
# write a logic, focus on calling and concept) 
# • A Pipeline class that: 
# o Accepts any model 
# o Uses composition to chain transformations 
# o Overloads __call__() to run predictions 
# • Encapsulates internal steps 
from abc import ABC,abstractmethod
class Model(ABC):
    @abstractmethod
    def train(self,data):
        pass
    @abstractmethod
    def predict(self,data):
        pass

class LinearRegressionModel(Model):
    def train(self, data):
        print("Training Linear Regression model")
    def predict(self,data):
        return "Linear Regression prediction"
class DecisionTreeModel(Model):
    def train(self, data):
        print("Training Decision Tree model")
    def predict(self, data):
        return "Decision Tree Prediction"

class Pipeline:
    def __init__(self,model,steps):
        self.model = model
        self.__steps = steps.copy()
    def train(self,data):
        for step in self.__steps:
            data = step(data)
        self.model.train(data)         
    def __call__(self,data):
        for step in self.__steps:
            data = step(data)
        return self.model.predict(data)

# Transformatin functions
def remove_missing(data):
    print("Removing missing values")
    return data
def scale_data(data):
    print("Scaling data")
    return data
# create a pipeline
model = LinearRegressionModel()
pipeline = Pipeline(
    model,
    [remove_missing,scale_data]
    )
# Train the model
pipeline.train([10,20,30])

print("\nPrediction:")
print(pipeline([40,50]))
```

```python
# 15. Create a mini version of Amazon with: 
# Classes: 
# • User 
# • Product 
# • Seller(User) 
# • Buyer(User) 
# • Order 
# • Cart 
# Requirements (must use all OOP concepts): 
# >Inheritance: Seller and Buyer extend User 
# >Encapsulation: protect internal cart list, user password 
# >Abstraction: base class User defines abstract get_role() 
# >Polymorphism: different users behave differently in checkout 
# >Composition: Buyer “has” a Cart 
# >Operator overloading: 
# • + to add product to Cart 
# • - to remove product 
# >Properties: validate product price 
# >Class methods: tracking total users 
# >Static methods: validating product IDs 
# >__str__ for readable summaries 
# >MRO behavior when Buyer inherits from multiple mixins (e.g., RewardsMixin) 

from abc import ABC,abstractmethod

# Product class
class Product:
    def __init__(self,product_id,name,price):
        if(not self.validate_id(product_id)):
            raise ValueError("Invalid product ID")
        self.product_id = product_id
        self.name = name
        self.price = price
    @staticmethod
    def validate_id(product_id):
        return isinstance(product_id,int) and product_id>0
    @property
    def price(self):
        return self.__price
    @price.setter
    def price(self,value):
        if value<0:
            raise ValueError("Price cannot be negative")
        self.__price = value
    def __str__(self):
        return f"{self.name} - Rs.{self.price}"

# cart class
class Cart:
    def __init__(self):
        self.__products = []
    def __add__(self, product):
        self.__products.append(product)
        return self
    def __sub__(self, product):
        if(product in self.__products):
            self.__products.remove(product)
        return self
    def total(self):
        amount = 0
        for product in self.__products:
            amount += product.price
        return amount
    def __str__(self):
        result = "Cart:\n"
        for product in self.__products:
            result += str(product) + "\n"
        result += "Total: Rs." + str(self.total())
        return result

# Abstract User class
class User(ABC):
    total_users = 0
    def __init__(self,name,password):
        self.name = name
        self.__password = password
        User.total_users += 1
    @abstractmethod
    def get_role(self):
        pass
    def check_password(self,password):
        return self.__password == password
    @classmethod
    def get_total_users(cls):
        return User.total_users
    def checkout(self,cart):
        print(self.name,"is checking out as",self.get_role())

# Seller class
class Seller(User):
    def get_role(self):
        return "Seller"
    def checkout(self,cart):
        print("Seller can't purchase through seller checkout.")
# Rewards mixin
class RewardsMixin:
    def checkout(self,cart):
        print("RewardsMixin: Checking reward points")
        super().checkout(cart)
# Buyer class
class Buyer(RewardsMixin,User):
    def __init__(self, name, password):
        super().__init__(name, password)
        self.cart = Cart()
    def get_role(self):
        return "Buyer"
    def checkout(self):
        print("Buyer checkout started")
        print(self.cart)
        print("Order placed successfully!")
    def __str__(self):
        return f"Buyer: {self.name}"

# Order class
class Order:
    def __init__(self,buyer):
        self.buyer = buyer
        self.amount = buyer.cart.total()
    def __str__(self):
        return (
            f"Order by {self.buyer.name},"
            f"Total: Rs.{self.amount}"
        )

# create products
p1 = Product(1,"Laptop",50000)
p2 = Product(2,"Mouse",500)
p3 = Product(3,"Keyboard",1000)

# create users
buyer = Buyer("Akhil","pass123")
seller = Seller("Ravi","seller123")

# add products using +
buyer.cart = buyer.cart + p1
buyer.cart = buyer.cart + p2
buyer.cart = buyer.cart + p3

# remove product using -
buyer.cart = buyer.cart - p3

# display products and cart
print(p1)
print(buyer.cart)

# password validation
print("Correct password:",buyer.check_password("pass123"))

# different user checkout behavior
seller.checkout(buyer.cart)
buyer.checkout()

# demonstrate mro behavior
print("\nMRO:")
for cls in Buyer.mro():
    print(cls.__name__)

# demonstrate class method
print("\nTotal users:",User.get_total_users())

# create order
order = Order(buyer)
print(order)
```
