```python
# Inheritance: Acquiring properties from one class to another class.
# 1. Single-level inheritance
# Single parent class - single child class
# Single base class - single derived class
# Single super class - single sub class
# A     -> super/base/parent class  (A)
# |
# ⬇️
# B     -> sub/derived/child class  (A+B)
# 'B' is derived from 'A'
print("1.Single-level Inheritance:")
class A:
    pass
class B:
    pass
# single level inheritance
class A:
    pass
class B(A):
    pass
obj = B()
print(type(obj))
print()
class A:
    def m1(self):
        print("This is m1.")
class B(A):
    pass
obj1 = B()
obj1.m1()
```

```python
# 2. Multi-level Inheritance: 
# One parent class & one child but in btw we have intermediate classes
# A     -> super/base/parent class      (A)
# |
# ⬇️
# B     ->intermediate class            (A+B)   
# |
# ⬇️
# C     -> sub/derived/child class      (A+B+C)
print("2.Multi-level Inheritance:")
class A:
    def m1(self):
        print("This is m1.")
class B(A):
    def m2(self):
        print("This is m2.")
class C(B):
    pass
obj1 = C()
obj1.m1()
obj1.m2()
```

```python
# 3. Multiple Inheritance:
# One child class takes properties/inherits properties of multiple parent classes.
# One child class - multiple parent class
# A(A)    B(B)
#  |       |
#  ↘️     ↙️
#    C(A+B)
# Order matters in class C and execution works in that order
# First searches in class A then searches in class B
print("3. Multiple Inheritance:")
class A:
    def m1(self):
        print("This is m1.")
class B:
    def m2(self):
        print("This is m2.")
class C(A,B):
    pass
obj1 = C()
obj1.m1()
obj1.m2()
```

```python
# 4. Hierarchial Inheritance:
# single parent class ->(properties)-> multiple child classes
print("4. Hierarchial Inheritance:")
class A:
    def m1(self):
        print("This is m1.")
class B(A):
    def m2(self):
        print("This is m2.")
class C(A):
    def m3(self):
        print("This is m3.")
class D(A):
    def m4(self):
        print("This is m4.")
obj1 = B()
obj1.m1()
obj1.m2()
print()
obj2 = C()
obj2.m1()
obj2.m3()
print()
obj3 = D()
obj3.m1()
obj3.m4()
```

```python
# 5. Hybrid Inheritance:
# combination of more than 2 inheritances
#      A(A)
#       |
#   ↙️    ↘️
# B(A+B)  C(A+C)
#  |        |
#  ↘️      ↙️    
#   D(A+B+C)
class A:
    def m1(self):
        print("This is m1.")
class B(A):
    def m2(self):
        print("This is m2.")
class C(A):
    def m3(self):
        print("This is m3.")
class D(B,C):
    pass
obj1 = D()
obj1.m1()
obj1.m2()
obj1.m3()
```

```python
# 10. ATM System Using Multiple Inheritance 
# Class 1: SBI 
# • Create the methods deposit(amount) and check_balance().  
# Class 2: UnionBank 
# • Create the methods withdraw(amount) and mini_statement().  
# Class 3: ATM (inherits SBI and UnionBank) 
# • Create the methods menu() and transaction().  
# • Allow the user to perform banking operations.  
# Driver Code 
# • Create an object of the ATM class.  
# • Call the transaction() method.
class SBI:
    def __init__(self,balance):
        self.balance = balance
        self.trans = []
    def deposit(self,amount):
        if(amount>0):
            self.balance += amount
            self.trans.append("Deposited: "+str(amount)) 
            print("Your account as credited:",amount)
        else:
            print("Invalid Deposit") 
    def check_balance(self):
        print("Your current balance:",self.balance)
class UnionBank:
    def withdraw(self,amount):
        if(amount <= self.balance):
            self.balance -= amount
            self.trans.append("Withdrawn: "+str(amount))
            print("Your account as debited:",amount)
        else:
            if(amount>self.balance):
                print("Insufficient Funds")
            else:
                print("Enter a Valid amount to Withdraw")
    def mini_statement(self):
        print("Mini Statement:")
        for i in self.trans:
            print(i)
        print("Balance:",self.balance)
class ATM(SBI,UnionBank):
    def __init__(self, balance):
        super().__init__(balance)
    def menu(self):
        print("----ATM menu---")
        print("1.Check balance")
        print("2.Deposit")
        print("3.Withdraw")
        print("4.Mini-statement")
        print("5.Exit")
    def transaction(self):
        self.menu()
        while(True):
            choice = int(input("Enter your choice:"))
            if(choice==1):
                self.check_balance()
            elif(choice==2):
                amount = int(input("Enter amount to deposit:"))
                self.deposit(amount)
            elif(choice==3):
                amount = int(input("Enter amount to withdraw:"))
                self.withdraw(amount)
            elif(choice==4):
                self.mini_statement()
            else:
                print("ThankYou!")
                break
a = ATM(1000)
a.transaction()
```

```python
# 11. Paytm Application Using Multiple Inheritance 
# Write a Python program to implement a Paytm Application using multiple 
# inheritance. 
# Class 1: MobileRecharge 
# • Create the methods recharge_plans() and mobile_recharge().  
# Class 2: BusTicketBooking 
# • Create the methods display_buses() and book_ticket().  
# Class 3: ElectricityBills 
# • Create the methods bill_details() and pay_bill().  
# Class 4: Paytm (inherits MobileRecharge, BusTicketBooking, and 
# ElectricityBills) 
# Create the following methods: 
# • menu() – Display the available services.  
# • services() – Allow the user to choose and use any service (Mobile 
# Recharge, Bus Ticket Booking, or Electricity Bill Payment).
class MobileRecharge:
    def recharge_plans(self):
        print("-----Recharge Plans---")
        print("1. Rs.199 - 1.5GB/day")
        print("2. Rs.299 - 2GB/day")
        print("3. Rs.499 - 2.5GB/day")
    def mobile_recharge(self):
        Mobile_no = int(input("Enter your 10 digit mobile no.:"))
        print("Your Mobile No.:",Mobile_no)
        choice = int(input("Enter your recharge plan amount(1/2/3):"))
        if(choice==1):
            return 199
        elif(choice==2):
            return 299
        elif(choice==3):
            return 499
        else:
            return 0
class BusTicketBooking:
    def display_buses(self):
        print("----Avaiable Buses----")
        print("1.Hyderabad to Thirupathi - Rs.1000")
        print("2.Thirupathi to Hyderabad - Rs.800")
        print("3.Vizag to Hyderabad - Rs.900")
        print("4.Hyderabad to Vizag - Rs.1200")
    def book_tickets(self):
        choice = int(input("Enter your route no.(1/2/3/4):"))
        if(choice==1):
            return 1000
        elif(choice==2):
            return 800
        elif(choice==3):
            return 900
        elif(choice==4):
            return 1200
        else:
            return 0
class EletricityBills:
    def bill_details(self):
        print("1. 0-100 units = Rs.2/per unit")
        print("2. 100-200 units = Rs.3/per unit")
        print("3. Above 200 units = Rs.5/per unit")
    def pay_bills(self):
        choice = int(input("Enter no. of units(1/2/3):"))
        if(choice==1):
            return choice*2
        elif(choice==2):
            return choice*3
        elif(choice==3):
            return choice*5
        else:
            return 0
class Paytm(MobileRecharge,BusTicketBooking,EletricityBills):
    def menu(self):
        print("====Paytm Menu====")
        print("1. Mobile Recharge")
        print("2. Bus Ticket Booking")
        print("3. Electricity Bills")
        print("4. Exit")
    def services(self):
        while(True):
            self.menu()
            choice = int(input("Choose service(1/2/3):"))
            if(choice==1):
                self.recharge_plans()
                price = self.mobile_recharge()
                if(price==0):
                    print("Invalid Recharge plan")
                else:
                    print("Recharge Bill:",price)
            elif(choice==2):
                self.display_buses()
                price = self.book_tickets()
                if(price==0):
                    print("Invalid Bus ticket choosen")
                else:
                    print("Your Bus Ticket Price:",price)
            elif(choice==3):
                self.bill_details()
                price = self.pay_bills()
                print("Your Eletricity Bill:",price)
            elif(choice>=4):
                print("Thankyou for using Paytm")
                break
p = Paytm()
p.services()
```
