```python
def sum_recursive(first, *rest):
    # Base case: if 'rest' is empty, return the last number
    if not rest:
        return first
    
    # Recursive case: add first to the sum of the rest
    return first + sum_recursive(*rest)

# Example usage
print(sum_recursive(1, 2, 3, 4, 5))  # Output: 15   
```

```python
# Q8.  FULL PIPELINE: Build a mini data pipeline. Start with a list of student dictionaries [{name, score}]. Use filter() to keep scores >= 60, map() to add a 'grade' key ('Pass'), and sorted() to sort by score descending. Print the final result.
students = [
    {"name":'abhilash','score':90},
    {'name':'akhil','score':50},
    {'name':'vamshi','score':78}
] 
filtered_stu = list(filter(lambda x: x['score']>60,students))
mapped_stu = list(map(lambda x: dict(x,grade='pass'),filtered_stu))
result = sorted(mapped_stu,key = lambda x:x['score'],reverse=True)
print(result)
```

```python
students = [
    {"name":'abhilash','score':90},
    {'name':'akhil','score':50},
    {'name':'vamshi','score':78}
]
stu = list(filter(lambda x:x['score']>60, students))
stu = list(map(lambda x:{**x, 'grade':'pass'},stu))
stu = sorted(stu,key= lambda x: x['score'],reverse=True)
print(stu)
```

```python
# Q9.  LAMBDA + sorted() + FUNCTION REFERENCE: Store three sort strategies in a dictionary: by_name, by_score, by_length. Let the user choose a strategy by name, then apply it to sort a list of tuples. 
list = [
    ('Abhilash',90),
    ('Raju',78),
    ('Harikrishna',70),
    ('Manisai',67)
]
strategies = {
    'by_name' : lambda x:x[0],
    'by_score' : lambda x:x[1],
    'by_length' : lambda x:len(x[0])
}

choice = input("Enter your choice(by_name/by_score/by_length): ")

print(sorted(list,key= strategies[choice]))
```

```python
# Q10.  ALL CONCEPTS: Write a function calculator(*args, operation='add', **options) that: 
# (a) uses *args to collect numbers, 
# (b) uses a default 'add' operation, 
# (c) supports operations: 'add', 'multiply', 'max', 'min' using a dict of lambda functions, 
# (d) if options contains show_steps=True, prints each step of the calculation. 
def calculator(*args, operation = 'add', **options):
    op = {
        'add': lambda x,y : x+y,
        'mul': lambda x,y : x*y,
        'max': lambda x,y: x if x>y else y,
        'min': lambda x,y : x if x<y else y
    }
    func = op[operation]
    res = args[0] # if only one value is given as input
    for i in args[1:]: # first value is assigned to res (res = args[0]) so we can do from next index 1 to end so we do args[1:]
        # func (res,i)
        if options.get('show_steps') == True:
            print(res,i,operation,":",func(res,i))
        res = func(res,i)
    return res

print("One value:")
print(calculator(1))

print("\nWithout steps")
print("Addition:")
print(calculator(2,8,18,11))
print("\nMultiplication:")
print(calculator(2,8,18,11,operation='mul'))
print("\nMax:")
print(calculator(2,8,18,11,operation='max'))
print("\nMin:")
print(calculator(2,8,18,11,operation='min'))

print("\nWith show steps:")

print("\nAddition:")
print(calculator(2,8,18,11,show_steps= True))

print("\nmultiplication:")
print(calculator(2,8,18,11,operation='mul',show_steps= True))

print("\nMax:")
print(calculator(2,8,18,11,operation='max',show_steps= True))

print("\nMin:")
print(calculator(2,8,18,11,operation='min',show_steps= True))
```
