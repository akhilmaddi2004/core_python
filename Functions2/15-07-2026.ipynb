```python
# 1.
# Write a python program that creates a nested list and perform both shallow copy and deep copy operations.
# Modify an inner element of the original list and display all lists.
# Based on the output, explain the difference between shallow copy and deep copy.

org_list = [['python','java','c'],['html','css','js']]
print("Original list:",org_list)

import copy

print("\nOriginal list before shallow copy:",org_list)
# shallow copy 
sh_cpy = copy.copy(org_list)
sh_cpy[1][1] = 'react.js'
print("Shallow copy of original copy after shallowing:",sh_cpy)
print("Original copy after shallowing:",org_list)

# deepcopy
print("\nOriginal list:",org_list)

dp_cpy = copy.deepcopy(org_list)
dp_cpy[0][1] = "C++"
print("Deep copy of original copy after deep copy:",dp_cpy)
print("Original copy after deep copy:",org_list)

# shallow copy - if any changes happen in shallow copy then it will effect on original copy as well.
# deepcopy - if any changes happen in deep copy no changes occur in original copy
```

```python
org_list = [[1,2,3],[4,5,6],[7,8,9]]

print("Original list: ",org_list)

import copy

# shallow copy
sc = copy.copy(org_list)
sc[1][0] = 10
print("org list after shallowing:",org_list)
print("Shallow copied list:",sc)

# deep copy
dp = copy.deepcopy(org_list)
dp[2][1] = 11
print("org list after deep copy:",org_list)
print("deep copied list:",dp)
```

```python
# 2.
# Define a function calculate_total(price,tax=5,discount=0) that calculates the final price after adding tax subtracting discount.
# Demonstrate the function with different function calls to show how default arguments work.

def calculate_total(price,tax=5,discount=0):
    total = price+tax
    final_price = total-discount
    return  final_price
print("final price with positional arguments:",calculate_total(100)) # calling function with only required positional arguments
print("final price with all keyword arguments:",calculate_total(price= 100,tax = 10,discount=5)) # calling function with keyword arguments
print("final price:",calculate_total(price=100))
print("final price:",calculate_total(100,tax=10,discount=20))
print("final price:",calculate_total(100,tax = 5))
print("final price:",calculate_total(100,discount = 5))
```

```python
# 3.
# Write a function create_profile(name,*skills,**details) that accepts a person's name, multiple skills, and
# additional details such as age,city, and experience.
# Display the profile in a structured format.
def create_profile(name,*skills,**details):
    print("Profile Summary:")
    print("name:",name)
    print("Skills:")
    for skill in skills:
        print(skill,end=",")
    print("\nAdditional Details:")
    for detail,about in details.items():
        print(f"{detail}:{about}")

create_profile("Akhil",
               "Python","SQL","HTML","CSS","JS",
               age = 22, city = "Hyderabad",experience = 0)
```

```python
# 4.
# create functions for addition, subtraction, and multiplication.
# store these functions in a dictionary and write a program that takes user input to select an operation and
# performs it dynamically using function references.

# addition function
def addition(a,b):
    return a+b

# subtraction function
def subtraction(a,b):
    return a-b

# multiplication function
def multiplication(a,b):
    return a*b

# storing all three functions in dictionary using referencing
dict = {
    "add":addition,
    "sub":subtraction,
    "mul":multiplication
}

# calling dynamically using function references
op = input("Enter an operation(add,sub,mul):")
print(dict[op](20,5))

# calling inputs dynamically
op1 = input("Enter an operation(add,sub,mul):")
a = int(input("Enter your first number:"))
b = int(input("Enter your second number:"))
print(dict[op1](a,b))
```
