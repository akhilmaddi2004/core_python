```python
def swiggy(customer_name,order_type="regular",*food_items,**details):
    print("Hi!",customer_name)
    print("your order type is:",order_type)
    print("List of ordered items:")
    total_bill = 0
    for item in food_items:
        print(item[0],":Rs.",item[1])
        total_bill += item[1]
    if total_bill>300:
        total_bill = total_bill - (total_bill*0.1)
        print("Your total bill is:",total_bill)
    else:
        print("Your total bill is:",total_bill)
    print("Total items in cart:",len(food_items))
    print("Additional details:")
    for detail,about in details.items():
        print(detail,":",about)
swiggy("Akhil","Swiggy One",
       ["Biryani", 300],["Coke",50],["Sweet", 30],
       payment_mode = "UPI", delivery_instructions = "Don't ring the Bell")
```

## function referencing


```python
# In python, functions are first-class objects, meaning they can be treated like any other variable.

# A function reference allows you to refer to a function without immediately calling it, enabling powerful programming techniques such as passing functions as arguments, storing them in variables, and returning them from other functions.

# 1. Assignment: Functions can be assigned to a variable 
# (eg. my_func = original_function).

# 2. Argument passing: Functions can be passed as parameters to other functions.
# (eg. process(callback_function)).

# 3. Return values: Functions can be returned from other functions, enabling closures and factories.

# 4. Storage: Functions can be stored in data structures like lists or dictionaries
# (eg. function_dict = {'add':add_func})
```

```python
# Function referencing: refering a function in a variable
zomato = swiggy
zomato("Abhilash","Zomato Gold",
       ["Shawarma",100],["Chicken roll",150],["Coke",50],["Sweet",50],
       payment_mode="UPI",delivery_instructions= "Arrive fast")
```

```python
# we have three types of function referencing
# 1. one function is stored in a variable and it gets same features of original feature
# 2. one function is stored in dictionary, list
# 3. one function passed in another function as parameter

# 1. function stored in variable
def say_hi(name):
    print("Hi!",name)
greet = say_hi

greet("Akhil")
```

```python
# 2. one function is stored in dictionary,list
def add(a,b):
    return a+b
def sub(a,b):
    return a-b
def mul(a,b):
    return a*b
calculate = [add,sub,mul]
print("Addition:",calculate[0](2,3))
print("Subtaction",calculate[1](10,5))
print("Multiply",calculate[2](23,2))
```

```python
# dict
operations ={
    '+':add,
    '-':sub,
    '*':mul 
}
print(operations['+'](10,20))
op = input("Enter operation:")
print(operations[op](10,20))
```

```python
# 3. one function is passed as a parameter to another function
def square(a):
    return a**2
def cube(a):
    return a ** 3
def double(func,value):
    return func(func(value))
def d(func,value):
    return func(value)
print(d(square,2))
print(double(square,4))
print(double(cube,2))
```

```python
# 3. Store the functions upper,lower,and title (string methods) in a dictionary. Let the user choose which one to apply
def upper(a):
    return a.upper()
def lower(a):
    return a.lower()
def title(a):
    return a.title()
str_methods={
    "upper":upper,
    "lower":lower,
    "title":title
}
print(str_methods["lower"]("Abhilash"))
method = input("Enter a str_method(upper,lower,title):")
input_str = input("Enter your string")
print(str_methods[method](input_str))
```
