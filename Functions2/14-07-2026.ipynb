```python
# functional referencing
# 1. function is assigned a variable
def greet(name):
    print("Hi!",name)
x = greet
# x stores greet functions print stmt or return stmt and 
# x acts same as greet function will get same features
greet("akhil")
x("akhil")
```

```python
def greet(name):
    print(name)
x = greet
x("Akhil")


a = 10
def add(a,b):
    print(a+b)
add(a,10)

# next passing functions to a list because funcitons are objects
l = [10,20,20,add,greet]
l[4]("Akhil")
l[3](10,20)
```

```python
# we can't add the function to another function but we can call the function
def greet(name):
    print(name)
b = greet

def m1(a,b):
    print("The value of a is ",a)
    print("The value of b is ",b)
m1(a,b)

print(greet)
print(m1)
```

```python
# shallow copy and deepcopy
# shallow copy - changes occur in both original and copied copy
import copy
l = [[10,20,30],[40,50,60]]
print("list 'l' before shallow copy:",l)
sc = copy.copy(l)
sc[0][0] = 34
print("list 'l' after shallow copy:",l)
print("'l','sc' after shallow copy:",sc)
```

```python
# deepcopy - changes occur only in copied copy not in occur in original copy
l = [[1,2,3],[4,5,6]]
print("Original list before deepcopy:",l)
dc = copy.deepcopy(l)
dc[0][0] = 10
print("Original list after deepcopy:",l)
print("Deepcopied list after deepcopy:",dc)
```
