```python
def dec1(func):
    def wrap1(*args,**kwargs):
        print("Hi")
        func(*args,**kwargs)
    return wrap1
def my_name(name):
    print("My name is",name)
x = dec1(my_name)
x("Akhil")
```

```python
def dec2(func):
    def wrap2(*args,**kwargs):
        func(*args,**kwargs)
        print("Bye")
    return wrap2
def dec1(func):
    def wrap1(*args,**kwargs):
        print("Hi")
        func(*args,**kwargs)
    return wrap1
def my_name(name):
    print("My name is",name)
x = dec1(my_name)
y = dec2(x)
y("Akhil")

# single line calling
d = dec2(dec1(my_name))
d("Arjun")
```

```python
def dec2(func):
    def wrap2(*args,**kwargs):
        func(*args,**kwargs)
        print("Bye")
    return wrap2
def dec1(func):
    def wrap1(*args,**kwargs):
        print("Hi")
        func(*args,**kwargs)
    return wrap1
@dec2
@dec1
def my_name(name):
    print("My name is",name)
my_name("Akhil")
```

```python
# The execution of decorators occurs in bottom-top approach

# Firstly we will give my_name func to dec1 then after completing or returning whole dec1 function, then that return output given to dec2 so decorators execution is bottom to top
```

```python
# A banking application has a function check_balance(). Create two decorators: verify_user, which prints "User verified", and log_transaction, which prints "Transaction logged". Apply both decorators to check_balance() and display "Balance displayed" from the original function.
def verify_user(func):
    def wrap1(*args,**kwargs):
        print("User verified")
        func(*args,**kwargs)
    return wrap1
def log_transaction(func):
    def wrap2(*args,**kwargs):
        print("Transaction logged")
        func(*args,**kwargs)
    return wrap2
@verify_user
@log_transaction
def check_balance():
    print("Balance displayed")
check_balance()
```

```python
def verify_user(func):
    def wrap1(*args,**kwargs):
        print("User verified")
        func(*args,**kwargs)
    return wrap1
def log_transaction(func):
    def wrap2(*args,**kwargs):
        print("Transaction logged")
        func(*args,**kwargs)
    return wrap2
def check_balance():
    print("Balance displayed")
check_balance = verify_user(log_transaction(check_balance))
check_balance()
```

```python
# An online examination system has a function start_exam(student). Before allowing the student to start the exam, the system must verify the student’s login and then log the exam activity. Create two decorators, login_required and log_activity, and apply both decorators to start_exam(). The function should finally display "Exam started for <student>".
def login_required(func):
    def wrap1(*args,**kwargs):
        print("Login verified")
        func(*args,**kwargs)
    return wrap1
def login_activity(func):
    def wrap2(*args,**kwargs):
        print("Exam activity logged")
        func(*args,**kwargs)
    return wrap2
@login_required
@login_activity
def start_exam(student):
    print(f"Exam started for <{student}>")
start_exam("Akhil")
```

```python
def login_required(func):
    def wrap1(*args,**kwargs):
        print("Login verified")
        func(*args,**kwargs)
    return wrap1
def login_activity(func):
    def wrap2(*args,**kwargs):
        print("Exam activity logged")
        func(*args,**kwargs)
    return wrap2
def start_exam(student):
    print(f"Exam started for {student}")
start_exam = login_required(login_activity(start_exam))
start_exam("Akhil")
```
